```python
# ============================================================
# CELL 2 — GENERAL HELPERS
# ============================================================

def find_all(filename):
    return sorted(
        KAGGLE_INPUT.rglob(filename),
        key=lambda p: (len(p.parts), str(p))
    )

def normalize_probabilities(probabilities):
    p = np.asarray(probabilities, dtype=np.float64)

    if p.ndim != 2 or p.shape[1] != NUM_CLASSES:
        raise ValueError(
            f"Expected probability shape (N,{NUM_CLASSES}), got {p.shape}"
        )

    p = np.nan_to_num(
        p,
        nan=0.0,
        posinf=0.0,
        neginf=0.0,
    )

    p = np.clip(p, 1e-12, None)
    p /= p.sum(axis=1, keepdims=True)

    return p.astype(np.float32)

def probability_columns(df):
    cols = [
        c for c in df.columns
        if c.startswith("prob_class_")
        and c.rsplit("_", 1)[-1].isdigit()
    ]
    if len(cols) == NUM_CLASSES:
        return sorted(
            cols,
            key=lambda c: int(c.rsplit("_", 1)[-1])
        )

    cols = [
        c for c in df.columns
        if c.startswith("prob_")
        and len(c.split("_")) >= 2
        and c.split("_")[1].isdigit()
    ]
    if len(cols) == NUM_CLASSES:
        return sorted(
            cols,
            key=lambda c: int(c.split("_")[1])
        )

    exact = [
        f"prob_{i}"
        for i in range(NUM_CLASSES)
        if f"prob_{i}" in df.columns
    ]
    if len(exact) == NUM_CLASSES:
        return exact

    raise ValueError(
        "Could not detect exactly 8 probability columns."
    )

def label_column(df):
    for c in ["y_true", "true_label", "label"]:
        if c in df.columns:
            return c

    raise ValueError(
        "True-label column not found."
    )

def load_prediction_csv(path):
    df = pd.read_csv(path)

    y_col = label_column(df)
    p_cols = probability_columns(df)

    y = df[y_col].to_numpy(np.int32)

    p = normalize_probabilities(
        df[p_cols].to_numpy(np.float32)
    )

    return y, p

def l2_normalize_rows(x, eps=1e-12):
    x = np.asarray(x, dtype=np.float64)

    norms = np.linalg.norm(
        x,
        axis=1,
        keepdims=True,
    )

    return (
        x
        / np.clip(norms, eps, None)
    ).astype(np.float32)

print("Helpers ready.")
```

```python
# ============================================================
# CELL 6 — LINEAR CKA
# ============================================================

def center_gram(k):
    k = np.asarray(k, dtype=np.float64)

    row_mean = k.mean(
        axis=1,
        keepdims=True,
    )

    col_mean = k.mean(
        axis=0,
        keepdims=True,
    )

    grand_mean = k.mean()

    return (
        k
        - row_mean
        - col_mean
        + grand_mean
    )

def linear_gram(features):
    x = np.asarray(
        features,
        dtype=np.float64,
    )

    x = x - x.mean(
        axis=0,
        keepdims=True,
    )

    return x @ x.T

def cka_from_centered_grams(kc, lc):
    numerator = np.sum(kc * lc)

    denominator = np.sqrt(
        np.sum(kc * kc)
        * np.sum(lc * lc)
    )

    if denominator <= 0:
        return np.nan

    return float(
        numerator / denominator
    )

def linear_cka(x, y):
    if len(x) != len(y):
        raise ValueError("CKA requires aligned samples.")

    k = center_gram(
        linear_gram(x)
    )

    l = center_gram(
        linear_gram(y)
    )

    return cka_from_centered_grams(
        k,
        l,
    )

REPRESENTATIONS = {
    "SimCLR": SIM_VAL_FEATURES,
    "Rotation": ROT_VAL_FEATURES,
    "DINOv2": DINO_VAL_FEATURES,
}

pair_names = [
    ("SimCLR", "Rotation"),
    ("SimCLR", "DINOv2"),
    ("Rotation", "DINOv2"),
]

cka_rows = []

for a, b in pair_names:
    score = linear_cka(
        REPRESENTATIONS[a],
        REPRESENTATIONS[b],
    )

    cka_rows.append({
        "representation_a": a,
        "representation_b": b,
        "linear_cka": score,
        "n_samples": len(Y_VAL),
        "split": "Validation",
    })

    print(
        f"{a} ↔ {b}: CKA = {score:.6f}"
    )

CKA_DF = pd.DataFrame(
    cka_rows
)

display(CKA_DF)

CKA_DF.to_csv(
    TABLE_DIR / "pairwise_linear_cka.csv",
    index=False,
)
```

```python
# ============================================================
# CELL 7 — CKA PERMUTATION SIGNIFICANCE TEST
# ============================================================

rng = np.random.default_rng(
    RANDOM_STATE
)

subset_n = min(
    CKA_PERMUTATION_SUBSET,
    len(Y_VAL),
)

# Stratified-ish fixed subset: sample approximately in proportion to class size.
selected = []

for c in range(NUM_CLASSES):
    idx = np.flatnonzero(
        Y_VAL == c
    )

    n_c = max(
        1,
        int(
            round(
                subset_n
                * len(idx)
                / len(Y_VAL)
            )
        )
    )

    n_c = min(
        n_c,
        len(idx),
    )

    selected.extend(
        rng.choice(
            idx,
            size=n_c,
            replace=False,
        ).tolist()
    )

selected = np.asarray(
    selected,
    dtype=np.int64,
)

# Enforce exact maximum size if rounding overshoots.
if len(selected) > subset_n:
    selected = rng.choice(
        selected,
        size=subset_n,
        replace=False,
    )

perm_rows = []

for a, b in pair_names:
    xa = REPRESENTATIONS[a][selected]
    xb = REPRESENTATIONS[b][selected]

    ka = center_gram(
        linear_gram(xa)
    )

    kb = center_gram(
        linear_gram(xb)
    )

    observed = cka_from_centered_grams(
        ka,
        kb,
    )

    null_scores = np.empty(
        CKA_PERMUTATIONS,
        dtype=np.float64,
    )

    n = len(selected)

    for r in range(
        CKA_PERMUTATIONS
    ):
        perm = rng.permutation(
            n
        )

        kb_perm = kb[
            np.ix_(
                perm,
                perm,
            )
        ]

        null_scores[r] = (
            cka_from_centered_grams(
                ka,
                kb_perm,
            )
        )

    p_value = (
        1
        + np.sum(
            null_scores
            >= observed
        )
    ) / (
        CKA_PERMUTATIONS
        + 1
    )

    perm_rows.append({
        "representation_a": a,
        "representation_b": b,
        "observed_cka_subset": observed,
        "null_mean": float(
            np.mean(
                null_scores
            )
        ),
        "null_std": float(
            np.std(
                null_scores,
                ddof=1,
            )
        ),
        "permutation_p": float(
            p_value
        ),
        "subset_n": int(
            len(selected)
        ),
        "permutations": CKA_PERMUTATIONS,
        "seed": RANDOM_STATE,
    })

    print(
        f"{a} ↔ {b}: "
        f"observed={observed:.5f}, "
        f"null={np.mean(null_scores):.5f}, "
        f"p={p_value:.5f}"
    )

CKA_PERM_DF = pd.DataFrame(
    perm_rows
)

display(CKA_PERM_DF)

CKA_PERM_DF.to_csv(
    TABLE_DIR / "cka_permutation_significance.csv",
    index=False,
)
```

```python
# ============================================================
# CELL 8 — REPRESENTATION QUALITY METRICS
# ============================================================

def reduce_for_quality(
    features,
    max_dim=QUALITY_PCA_DIM,
):
    x = np.asarray(
        features,
        dtype=np.float32,
    )

    x = StandardScaler(
        with_mean=True,
        with_std=True,
    ).fit_transform(x)

    dim = min(
        max_dim,
        x.shape[1],
        x.shape[0] - 1,
    )

    if dim < x.shape[1]:
        x = PCA(
            n_components=dim,
            random_state=RANDOM_STATE,
        ).fit_transform(x)

    return x.astype(
        np.float32
    )

def knn_label_purity(
    x,
    y,
    k=KNN_K,
):
    nn = NearestNeighbors(
        n_neighbors=k + 1,
        metric="euclidean",
    )

    nn.fit(x)

    indices = nn.kneighbors(
        x,
        return_distance=False,
    )[:, 1:]

    neighbor_labels = y[
        indices
    ]

    purity = np.mean(
        neighbor_labels
        == y[:, None]
    )

    return float(purity)

def fisher_ratio(
    x,
    y,
):
    global_mean = x.mean(
        axis=0
    )

    between = 0.0
    within = 0.0

    for c in range(
        NUM_CLASSES
    ):
        xc = x[
            y == c
        ]

        if len(xc) == 0:
            continue

        mean_c = xc.mean(
            axis=0
        )

        between += (
            len(xc)
            * np.sum(
                (
                    mean_c
                    - global_mean
                ) ** 2
            )
        )

        within += np.sum(
            (
                xc
                - mean_c
            ) ** 2
        )

    return float(
        between
        / max(
            within,
            1e-12,
        )
    )

quality_rows = []

for name, features in REPRESENTATIONS.items():
    xq = reduce_for_quality(
        features
    )

    silhouette = float(
        silhouette_score(
            xq,
            Y_VAL,
            metric="euclidean",
        )
    )

    purity = knn_label_purity(
        xq,
        Y_VAL,
        KNN_K,
    )

    fisher = fisher_ratio(
        xq,
        Y_VAL,
    )

    quality_rows.append({
        "representation": name,
        "silhouette": silhouette,
        "knn10_label_purity": purity,
        "fisher_ratio": fisher,
        "quality_dim": xq.shape[1],
        "split": "Validation",
    })

    print(
        name,
        "| silhouette",
        f"{silhouette:.4f}",
        "| kNN purity",
        f"{purity:.4f}",
        "| Fisher",
        f"{fisher:.4f}",
    )

QUALITY_DF = pd.DataFrame(
    quality_rows
)

display(QUALITY_DF)

QUALITY_DF.to_csv(
    TABLE_DIR / "representation_quality_metrics.csv",
    index=False,
)
```

```python
# ============================================================
# CELL 9 — UMAP VISUALIZATIONS
# ============================================================

def make_umap_embedding(
    features,
    model_name,
):
    x = np.asarray(
        features,
        dtype=np.float32,
    )

    scaler = StandardScaler()
    x_scaled = scaler.fit_transform(
        x
    )

    pca_dim = min(
        UMAP_PCA_DIM,
        x_scaled.shape[1],
        x_scaled.shape[0] - 1,
    )

    if pca_dim < x_scaled.shape[1]:
        x_reduced = PCA(
            n_components=pca_dim,
            random_state=RANDOM_STATE,
        ).fit_transform(
            x_scaled
        )
    else:
        x_reduced = x_scaled

    reducer = umap.UMAP(
        n_components=2,
        n_neighbors=30,
        min_dist=0.10,
        metric="euclidean",
        random_state=RANDOM_STATE,
    )

    emb = reducer.fit_transform(
        x_reduced
    )

    emb_df = pd.DataFrame({
        "x": emb[:, 0],
        "y": emb[:, 1],
        "label": Y_VAL,
        "class_name": [
            CLASS_NAMES[int(v)]
            for v in Y_VAL
        ],
        "representation": model_name,
    })

    emb_df.to_csv(
        EMBED_DIR
        / f"{model_name.lower()}_validation_umap.csv",
        index=False,
    )

    return emb_df

UMAP_RESULTS = {}

for model_name, features in REPRESENTATIONS.items():
    print("UMAP:", model_name)

    emb_df = make_umap_embedding(
        features,
        model_name,
    )

    UMAP_RESULTS[
        model_name
    ] = emb_df

    plt.figure(
        figsize=(7.4, 6.2)
    )

    for c, class_name in enumerate(
        CLASS_NAMES
    ):
        part = emb_df.loc[
            emb_df["label"] == c
        ]

        plt.scatter(
            part["x"],
            part["y"],
            s=8,
            alpha=0.65,
            label=class_name,
        )

    plt.xlabel("UMAP-1")
    plt.ylabel("UMAP-2")

    plt.title(
        f"{model_name} Representation — Original Validation Split"
    )

    plt.legend(
        markerscale=2,
        fontsize=8,
        ncol=2,
    )

    plt.tight_layout()

    plt.savefig(
        FIG_DIR
        / f"{model_name.lower()}_validation_umap.png",
        dpi=300,
        bbox_inches="tight",
    )

    plt.show()

gc.collect()
```

```python
# ============================================================
# CELL 10 — LOAD SAME-SPLIT PREDICTIONS
# ============================================================

# ---- Final proposed probability
final_hits = find_all(
    "final_internal_probabilities.npz"
)

FINAL_NPZ = None

for p in sorted(
    final_hits,
    key=lambda p: (
        0 if "reuse_only_all_model_ensemble" in str(p) else 9,
        len(p.parts),
        str(p),
    )
):
    try:
        with np.load(p) as z:
            if not {
                "y_test",
                "final_test_prob",
            }.issubset(z.files):
                continue

            yt = np.asarray(
                z["y_test"],
                dtype=np.int32,
            )

            if not np.array_equal(
                yt,
                Y_TEST,
            ):
                continue

        FINAL_NPZ = p
        break

    except Exception:
        continue

if FINAL_NPZ is None:
    raise FileNotFoundError(
        "Frozen final original-split probability file not found."
    )

with np.load(FINAL_NPZ) as z:
    FINAL_TEST_PROB = normalize_probabilities(
        z["final_test_prob"]
    )

# ---- Identify original pipeline root using SimCLR
sim_hits = find_all(
    "multiseed_probabilities_no_tta.npz"
)

SIMCLR_NPZ = None
PIPELINE_ROOT = None

for p in sim_hits:
    try:
        with np.load(p) as z:
            if not {
                "y_test",
                "test_prob",
            }.issubset(z.files):
                continue

            yt = np.asarray(
                z["y_test"],
                dtype=np.int32,
            )

            if not np.array_equal(
                yt,
                Y_TEST,
            ):
                continue

        SIMCLR_NPZ = p
        PIPELINE_ROOT = p.parent.parent
        break

    except Exception:
        continue

if SIMCLR_NPZ is None:
    raise FileNotFoundError(
        "Original SimCLR same-split probability file not found."
    )

with np.load(SIMCLR_NPZ) as z:
    SIMCLR_TEST_PROB = normalize_probabilities(
        z["test_prob"]
    )

# ---- Rotation
ROT_TEST_CSV = (
    PIPELINE_ROOT
    / "extended_imagenet_rotation_finetune"
    / "imagenet_rotation_test_predictions.csv"
)

rotation_y, ROTATION_TEST_PROB = load_prediction_csv(
    ROT_TEST_CSV
)

if not np.array_equal(
    rotation_y,
    Y_TEST,
):
    raise RuntimeError(
        "Rotation test labels are not aligned."
    )

# ---- DINO multiseed prediction, not representative seed
dino_prob_hits = find_all(
    "dinov2_original_split_multiseed_probabilities.npz"
)

DINO_PROB_NPZ = None

for p in dino_prob_hits:
    try:
        with np.load(p) as z:
            if not {
                "y_test",
                "test_prob",
            }.issubset(z.files):
                continue

            yt = np.asarray(
                z["y_test"],
                dtype=np.int32,
            )

            if not np.array_equal(
                yt,
                Y_TEST,
            ):
                continue

        DINO_PROB_NPZ = p
        break

    except Exception:
        continue

if DINO_PROB_NPZ is None:
    raise FileNotFoundError(
        "J2 DINO multiseed probability file not found."
    )

with np.load(DINO_PROB_NPZ) as z:
    DINO_TEST_PROB = normalize_probabilities(
        z["test_prob"]
    )

PREDICTIONS = {
    "SimCLR": np.argmax(
        SIMCLR_TEST_PROB,
        axis=1,
    ),
    "Rotation": np.argmax(
        ROTATION_TEST_PROB,
        axis=1,
    ),
    "DINOv2": np.argmax(
        DINO_TEST_PROB,
        axis=1,
    ),
    "Final Proposed": np.argmax(
        FINAL_TEST_PROB,
        axis=1,
    ),
}

print("Prediction sources loaded and aligned.")
```

```python
# ============================================================
# CELL 11 — PAIRWISE PREDICTION COMPLEMENTARITY
# ============================================================

def pair_complementarity(
    pred_a,
    pred_b,
    y_true,
):
    correct_a = (
        pred_a == y_true
    )

    correct_b = (
        pred_b == y_true
    )

    n11 = int(
        np.sum(
            correct_a
            & correct_b
        )
    )

    n10 = int(
        np.sum(
            correct_a
            & ~correct_b
        )
    )

    n01 = int(
        np.sum(
            ~correct_a
            & correct_b
        )
    )

    n00 = int(
        np.sum(
            ~correct_a
            & ~correct_b
        )
    )

    denominator = (
        n11 * n00
        + n10 * n01
    )

    q_stat = (
        np.nan
        if denominator == 0
        else (
            n11 * n00
            - n10 * n01
        ) / denominator
    )

    disagreement = float(
        np.mean(
            pred_a
            != pred_b
        )
    )

    double_fault = (
        n00
        / len(y_true)
    )

    return {
        "both_correct": n11,
        "a_only_correct": n10,
        "b_only_correct": n01,
        "both_wrong": n00,
        "disagreement_rate": disagreement,
        "q_statistic": float(q_stat),
        "double_fault_rate": float(double_fault),
    }

pairwise_rows = []

prediction_names = [
    "SimCLR",
    "Rotation",
    "DINOv2",
    "Final Proposed",
]

for i in range(
    len(prediction_names)
):
    for j in range(
        i + 1,
        len(prediction_names)
    ):
        a = prediction_names[i]
        b = prediction_names[j]

        stats = pair_complementarity(
            PREDICTIONS[a],
            PREDICTIONS[b],
            Y_TEST,
        )

        pairwise_rows.append({
            "model_a": a,
            "model_b": b,
            **stats,
        })

PAIRWISE_PRED_DF = pd.DataFrame(
    pairwise_rows
)

display(PAIRWISE_PRED_DF)

PAIRWISE_PRED_DF.to_csv(
    TABLE_DIR / "pairwise_prediction_complementarity.csv",
    index=False,
)
```

```python
# ============================================================
# CELL 12 — ERROR RESCUE BY THE FINAL MODEL
# ============================================================

final_pred = PREDICTIONS[
    "Final Proposed"
]

final_correct = (
    final_pred == Y_TEST
)

rescue_rows = []

for source_name in [
    "SimCLR",
    "Rotation",
    "DINOv2",
]:
    source_pred = PREDICTIONS[
        source_name
    ]

    source_correct = (
        source_pred == Y_TEST
    )

    source_errors = ~source_correct

    rescued = (
        source_errors
        & final_correct
    )

    lost = (
        source_correct
        & ~final_correct
    )

    rescue_rows.append({
        "source_model": source_name,
        "source_total_errors": int(
            np.sum(
                source_errors
            )
        ),
        "errors_rescued_by_final": int(
            np.sum(
                rescued
            )
        ),
        "rescue_rate_among_source_errors": float(
            np.sum(
                rescued
            )
            / max(
                np.sum(
                    source_errors
                ),
                1,
            )
        ),
        "source_correct_but_final_wrong": int(
            np.sum(
                lost
            )
        ),
        "net_correctness_gain": int(
            np.sum(
                rescued
            )
            - np.sum(
                lost
            )
        ),
    })

RESCUE_DF = pd.DataFrame(
    rescue_rows
)

display(RESCUE_DF)

RESCUE_DF.to_csv(
    TABLE_DIR / "final_model_error_rescue_summary.csv",
    index=False,
)
```

```python
# ============================================================
# CELL 13 — CLASS-WISE UNIQUE CORRECT / RESCUE ANALYSIS
# ============================================================

class_rows = []

for source_name in [
    "SimCLR",
    "Rotation",
    "DINOv2",
]:
    source_pred = PREDICTIONS[
        source_name
    ]

    source_correct = (
        source_pred == Y_TEST
    )

    for c, class_name in enumerate(
        CLASS_NAMES
    ):
        mask = (
            Y_TEST == c
        )

        source_unique_vs_final = int(
            np.sum(
                mask
                & source_correct
                & ~final_correct
            )
        )

        final_rescue = int(
            np.sum(
                mask
                & ~source_correct
                & final_correct
            )
        )

        class_rows.append({
            "source_model": source_name,
            "class_id": c,
            "class_name": class_name,
            "class_n": int(
                np.sum(mask)
            ),
            "source_correct_final_wrong": source_unique_vs_final,
            "source_wrong_final_correct": final_rescue,
            "net_final_rescue": (
                final_rescue
                - source_unique_vs_final
            ),
        })

CLASS_RESCUE_DF = pd.DataFrame(
    class_rows
)

display(CLASS_RESCUE_DF)

CLASS_RESCUE_DF.to_csv(
    TABLE_DIR / "classwise_final_error_rescue.csv",
    index=False,
)

# Rotation unique-correct against SimCLR specifically
sim_correct = (
    PREDICTIONS["SimCLR"]
    == Y_TEST
)

rot_correct = (
    PREDICTIONS["Rotation"]
    == Y_TEST
)

rotation_unique_rows = []

for c, class_name in enumerate(
    CLASS_NAMES
):
    mask = (
        Y_TEST == c
    )

    rotation_unique_rows.append({
        "class_id": c,
        "class_name": class_name,
        "class_n": int(
            np.sum(mask)
        ),
        "rotation_correct_simclr_wrong": int(
            np.sum(
                mask
                & rot_correct
                & ~sim_correct
            )
        ),
        "simclr_correct_rotation_wrong": int(
            np.sum(
                mask
                & sim_correct
                & ~rot_correct
            )
        ),
    })

ROTATION_UNIQUE_DF = pd.DataFrame(
    rotation_unique_rows
)

display(ROTATION_UNIQUE_DF)

ROTATION_UNIQUE_DF.to_csv(
    TABLE_DIR / "rotation_simclr_classwise_unique_correct.csv",
    index=False,
)
```
