# Reuse-Only Final All-Model Ensemble

This notebook **does not retrain Rotation, SimCLR, or any fusion model**.

It reuses the saved output of:

```text
last-part-of-the-previous-one
```

and loads the already-created validation/test probability files for:

- SimCLR multi-seed without TTA
- SimCLR multi-seed with validation-selected TTA
- Standalone Rotation
- FusionConcat
- FusionGated V2
- FusionGated V3
- Residual Fusion
- Previous fixed weighted ensemble, when available

It then performs:

1. Validation-only temperature calibration
2. Constrained all-model weighted search
3. Validation-fold logistic stacking
4. No-TTA SimCLR + stacking blend
5. Small validation-only class-bias correction
6. Frozen internal-test evaluation
7. Explicit comparison against single and fusion models

No original model-training cell is included.


````python
# ============================================================
# CELL 01: IMPORTS AND OUTPUT DIRECTORY
# ============================================================

import json
import math
import shutil
from collections import OrderedDict
from pathlib import Path

import joblib
import numpy as np
import pandas as pd

from IPython.display import display
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    log_loss,
    precision_score,
    recall_score,
    roc_auc_score,
)
from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, label_binarize

RANDOM_STATE = 42
RNG = np.random.default_rng(RANDOM_STATE)

KAGGLE_INPUT = Path("/kaggle/input")
OUTPUT_DIR = Path(
    "/kaggle/working/reuse_only_all_model_ensemble"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

PREFERRED_NOTEBOOK_SLUG = "last-part-of-the-previous-one"

CLASS_NAMES = [
    "MEL",
    "NV",
    "BCC",
    "AK",
    "BKL",
    "DF",
    "VASC",
    "SCC",
]
NUM_CLASSES = len(CLASS_NAMES)

print("=" * 88)
print("REUSE-ONLY ALL-MODEL ENSEMBLE")
print("=" * 88)
print("Output directory:", OUTPUT_DIR)
````

````python
# ============================================================
# CELL 02: LOCATE THE SAVED PREVIOUS-NOTEBOOK OUTPUT
# ============================================================

MARKER_RELATIVE_PATH = (
    "simclr_dominant_weighted_ensemble_v2/"
    "best_weights.json"
)

marker_matches = sorted(
    KAGGLE_INPUT.rglob(MARKER_RELATIVE_PATH)
)

if not marker_matches:
    raise FileNotFoundError(
        "The saved output of 'last-part-of-the-previous-one' "
        "was not found. Attach it through Kaggle Add Input."
    )

print("Framework markers found:")
for index, marker in enumerate(marker_matches):
    print(f"[{index}] {marker}")

preferred_matches = [
    marker
    for marker in marker_matches
    if PREFERRED_NOTEBOOK_SLUG in marker.parts
]

if not preferred_matches:
    raise FileNotFoundError(
        "Other framework outputs were found, but the required "
        "'last-part-of-the-previous-one' output was not attached."
    )

SELECTED_MARKER = sorted(
    preferred_matches,
    key=lambda path: (len(path.parts), str(path)),
)[0]

PIPELINE_ROOT = SELECTED_MARKER.parents[1]

print("\nSelected marker:")
print(SELECTED_MARKER)

print("\nPipeline root:")
print(PIPELINE_ROOT)
````

````python
# ============================================================
# CELL 03: LOAD ALL SAVED INTERNAL VALIDATION/TEST PROBABILITIES
# ============================================================

def normalize_probabilities(probabilities, epsilon=1e-12):
    probabilities = np.asarray(
        probabilities,
        dtype=np.float64,
    )

    if probabilities.ndim != 2:
        raise ValueError(
            "Probability arrays must be two-dimensional. "
            f"Received {probabilities.shape}."
        )

    probabilities = np.nan_to_num(
        probabilities,
        nan=0.0,
        posinf=0.0,
        neginf=0.0,
    )
    probabilities = np.clip(
        probabilities,
        epsilon,
        None,
    )
    probabilities /= probabilities.sum(
        axis=1,
        keepdims=True,
    )

    return probabilities.astype(np.float32)


def probability_columns(dataframe):
    standard_columns = [
        column
        for column in dataframe.columns
        if column.startswith("prob_class_")
    ]

    if standard_columns:
        return sorted(
            standard_columns,
            key=lambda column: int(
                column.rsplit("_", 1)[1]
            ),
        )

    rotation_columns = [
        column
        for column in dataframe.columns
        if column.startswith("prob_")
        and column.split("_")[1].isdigit()
    ]

    if rotation_columns:
        return sorted(
            rotation_columns,
            key=lambda column: int(
                column.split("_")[1]
            ),
        )

    raise ValueError(
        "No probability columns were found."
    )


def label_column(dataframe):
    for candidate in [
        "y_true",
        "true_label",
        "label",
    ]:
        if candidate in dataframe.columns:
            return candidate

    raise ValueError(
        "No true-label column was found."
    )


def load_prediction_csv(csv_path):
    dataframe = pd.read_csv(csv_path)

    label_name = label_column(dataframe)
    prob_columns = probability_columns(dataframe)

    y_true = dataframe[
        label_name
    ].to_numpy(dtype=np.int32)

    y_prob = normalize_probabilities(
        dataframe[
            prob_columns
        ].to_numpy(dtype=np.float32)
    )

    return y_true, y_prob


MODEL_PATHS = OrderedDict(
    [
        (
            "SimCLR_NoTTA",
            {
                "type": "npz",
                "path": (
                    PIPELINE_ROOT
                    / "simclr_finetune_multiseed_bacc_v2"
                    / "multiseed_probabilities_no_tta.npz"
                ),
            },
        ),
        (
            "SimCLR_SelectedTTA",
            {
                "type": "npz",
                "path": (
                    PIPELINE_ROOT
                    / "simclr_finetune_multiseed_bacc_v2"
                    / "tta_selection"
                    / "selected_multiseed_tta_probabilities.npz"
                ),
            },
        ),
        (
            "Rotation",
            {
                "type": "csv",
                "val": (
                    PIPELINE_ROOT
                    / "extended_imagenet_rotation_finetune"
                    / "imagenet_rotation_val_predictions.csv"
                ),
                "test": (
                    PIPELINE_ROOT
                    / "extended_imagenet_rotation_finetune"
                    / "imagenet_rotation_test_predictions.csv"
                ),
            },
        ),
        (
            "FusionConcat",
            {
                "type": "csv",
                "val": (
                    PIPELINE_ROOT
                    / "fusion_concat_multiseed_rep_v2"
                    / "full_evaluation"
                    / "val"
                    / "val_predictions.csv"
                ),
                "test": (
                    PIPELINE_ROOT
                    / "fusion_concat_multiseed_rep_v2"
                    / "full_evaluation"
                    / "test"
                    / "test_predictions.csv"
                ),
            },
        ),
        (
            "FusionGatedV2",
            {
                "type": "csv",
                "val": (
                    PIPELINE_ROOT
                    / "fusion_gated_v2_multiseed_rep_v2"
                    / "full_evaluation"
                    / "val"
                    / "val_predictions.csv"
                ),
                "test": (
                    PIPELINE_ROOT
                    / "fusion_gated_v2_multiseed_rep_v2"
                    / "full_evaluation"
                    / "test"
                    / "test_predictions.csv"
                ),
            },
        ),
        (
            "FusionGatedV3",
            {
                "type": "csv",
                "val": (
                    PIPELINE_ROOT
                    / "fusion_gated_v3_multiseed_rep_v2"
                    / "full_evaluation"
                    / "val"
                    / "val_predictions.csv"
                ),
                "test": (
                    PIPELINE_ROOT
                    / "fusion_gated_v3_multiseed_rep_v2"
                    / "full_evaluation"
                    / "test"
                    / "test_predictions.csv"
                ),
            },
        ),
        (
            "ResidualFusion",
            {
                "type": "npz",
                "path": (
                    PIPELINE_ROOT
                    / "residual_simclr_rotation_fusion_v1"
                    / "multiseed_probabilities.npz"
                ),
            },
        ),
    ]
)

VAL_PROBABILITIES = OrderedDict()
TEST_PROBABILITIES = OrderedDict()

Y_VAL = None
Y_TEST = None
availability_rows = []

for model_name, specification in MODEL_PATHS.items():
    available = False
    reason = ""

    try:
        if specification["type"] == "npz":
            artifact_path = specification["path"]

            if not artifact_path.exists():
                raise FileNotFoundError(
                    str(artifact_path)
                )

            with np.load(artifact_path) as saved:
                val_prob = normalize_probabilities(
                    saved["val_prob"]
                )
                test_prob = normalize_probabilities(
                    saved["test_prob"]
                )
                y_val = np.asarray(
                    saved["y_val"],
                    dtype=np.int32,
                )
                y_test = np.asarray(
                    saved["y_test"],
                    dtype=np.int32,
                )

        else:
            val_path = specification["val"]
            test_path = specification["test"]

            if not val_path.exists():
                raise FileNotFoundError(
                    str(val_path)
                )

            if not test_path.exists():
                raise FileNotFoundError(
                    str(test_path)
                )

            y_val, val_prob = load_prediction_csv(
                val_path
            )
            y_test, test_prob = load_prediction_csv(
                test_path
            )

        if Y_VAL is None:
            Y_VAL = y_val
            Y_TEST = y_test
        else:
            if not np.array_equal(
                Y_VAL,
                y_val,
            ):
                raise ValueError(
                    f"{model_name} validation labels are not aligned."
                )

            if not np.array_equal(
                Y_TEST,
                y_test,
            ):
                raise ValueError(
                    f"{model_name} test labels are not aligned."
                )

        VAL_PROBABILITIES[
            model_name
        ] = val_prob

        TEST_PROBABILITIES[
            model_name
        ] = test_prob

        available = True
        reason = "Loaded successfully"

    except Exception as error:
        reason = str(error)

    availability_rows.append(
        {
            "model": model_name,
            "available": available,
            "reason": reason,
        }
    )

AVAILABILITY_DF = pd.DataFrame(
    availability_rows
)

display(AVAILABILITY_DF)

required_models = [
    "SimCLR_NoTTA",
    "Rotation",
    "FusionConcat",
]

missing_required = [
    model_name
    for model_name in required_models
    if model_name not in VAL_PROBABILITIES
]

if missing_required:
    raise FileNotFoundError(
        "Required saved model predictions are missing: "
        f"{missing_required}"
    )

MODEL_NAMES = list(
    VAL_PROBABILITIES.keys()
)

anchor_shape = VAL_PROBABILITIES[
    "SimCLR_NoTTA"
].shape

for model_name in MODEL_NAMES:
    if (
        VAL_PROBABILITIES[
            model_name
        ].shape
        != anchor_shape
    ):
        raise ValueError(
            f"{model_name} validation shape does not match."
        )

    if (
        TEST_PROBABILITIES[
            model_name
        ].shape[1]
        != NUM_CLASSES
    ):
        raise ValueError(
            f"{model_name} class count does not match."
        )

print("\nLoaded models:")
print(MODEL_NAMES)

print("\nValidation shape:")
print(anchor_shape)

print("\nTest shape:")
print(
    TEST_PROBABILITIES[
        "SimCLR_NoTTA"
    ].shape
)

AVAILABILITY_DF.to_csv(
    OUTPUT_DIR
    / "loaded_saved_models.csv",
    index=False,
)
````

````python
# ============================================================
# CELL 04: METRICS AND VALIDATION-FOLD DEFINITIONS
# ============================================================

MODEL_CATEGORY = {
    "SimCLR_NoTTA": "Single",
    "SimCLR_SelectedTTA": "Single",
    "Rotation": "Single",
    "FusionConcat": "Fusion",
    "FusionGatedV2": "Fusion",
    "FusionGatedV3": "Fusion",
    "ResidualFusion": "Fusion",
}


def macro_specificity(confusion):
    total = confusion.sum()
    values = []

    for class_id in range(
        confusion.shape[0]
    ):
        true_positive = confusion[
            class_id,
            class_id,
        ]
        false_negative = (
            confusion[
                class_id,
                :,
            ].sum()
            - true_positive
        )
        false_positive = (
            confusion[
                :,
                class_id,
            ].sum()
            - true_positive
        )
        true_negative = (
            total
            - true_positive
            - false_negative
            - false_positive
        )

        denominator = (
            true_negative
            + false_positive
        )

        values.append(
            true_negative / denominator
            if denominator > 0
            else 0.0
        )

    return float(np.mean(values))


def evaluate_probabilities(
    y_true,
    y_prob,
    model_name,
    split_name,
    category,
):
    y_true = np.asarray(
        y_true,
        dtype=np.int32,
    )
    y_prob = normalize_probabilities(
        y_prob
    )
    y_pred = np.argmax(
        y_prob,
        axis=1,
    )

    confusion = confusion_matrix(
        y_true,
        y_pred,
        labels=np.arange(NUM_CLASSES),
    )

    y_binary = label_binarize(
        y_true,
        classes=np.arange(NUM_CLASSES),
    )

    try:
        auc = roc_auc_score(
            y_binary,
            y_prob,
            average="macro",
            multi_class="ovr",
        )
    except Exception:
        auc = float("nan")

    return {
        "model": model_name,
        "category": category,
        "split": split_name,
        "n_samples": int(
            len(y_true)
        ),
        "accuracy": float(
            accuracy_score(
                y_true,
                y_pred,
            )
        ),
        "bacc": float(
            balanced_accuracy_score(
                y_true,
                y_pred,
            )
        ),
        "macro_precision": float(
            precision_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0,
            )
        ),
        "macro_recall": float(
            recall_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0,
            )
        ),
        "macro_f1": float(
            f1_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0,
            )
        ),
        "weighted_f1": float(
            f1_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0,
            )
        ),
        "macro_specificity": (
            macro_specificity(
                confusion
            )
        ),
        "auc_macro_ovr": float(auc),
        "y_pred": y_pred,
        "confusion_matrix": confusion,
        "classification_report": (
            classification_report(
                y_true,
                y_pred,
                labels=np.arange(
                    NUM_CLASSES
                ),
                target_names=CLASS_NAMES,
                output_dict=True,
                zero_division=0,
            )
        ),
    }


FOLDS = list(
    StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=RANDOM_STATE,
    ).split(
        np.zeros(len(Y_VAL)),
        Y_VAL,
    )
)

ANCHOR_NAME = "SimCLR_NoTTA"
ANCHOR_VAL_PROB = VAL_PROBABILITIES[
    ANCHOR_NAME
]
ANCHOR_VAL_PRED = np.argmax(
    ANCHOR_VAL_PROB,
    axis=1,
)

ANCHOR_FOLD_BACC = np.asarray(
    [
        balanced_accuracy_score(
            Y_VAL[fold_indices],
            ANCHOR_VAL_PRED[
                fold_indices
            ],
        )
        for _, fold_indices
        in FOLDS
    ]
)

ANCHOR_FULL_BACC = float(
    balanced_accuracy_score(
        Y_VAL,
        ANCHOR_VAL_PRED,
    )
)

print(
    "No-TTA SimCLR validation BACC:",
    ANCHOR_FULL_BACC,
)
print(
    "Validation-fold BACCs:",
    ANCHOR_FOLD_BACC.tolist(),
)
````

````python
# ============================================================
# CELL 05: TEMPERATURE CALIBRATION USING VALIDATION ONLY
# ============================================================

TEMPERATURE_CANDIDATES = [
    0.60,
    0.70,
    0.80,
    0.90,
    1.00,
    1.10,
    1.20,
    1.30,
    1.50,
    1.75,
    2.00,
]


def temperature_scale(
    probabilities,
    temperature,
):
    probabilities = normalize_probabilities(
        probabilities
    )

    logits = np.log(
        np.clip(
            probabilities,
            1e-12,
            1.0,
        )
    )
    logits /= float(temperature)
    logits -= logits.max(
        axis=1,
        keepdims=True,
    )

    return normalize_probabilities(
        np.exp(logits)
    )


TEMPERATURES = {}
CALIBRATED_VAL = OrderedDict()
CALIBRATED_TEST = OrderedDict()
temperature_rows = []

for model_name in MODEL_NAMES:
    candidate_rows = []

    for temperature in TEMPERATURE_CANDIDATES:
        calibrated = temperature_scale(
            VAL_PROBABILITIES[
                model_name
            ],
            temperature,
        )

        candidate_rows.append(
            {
                "temperature": temperature,
                "nll": log_loss(
                    Y_VAL,
                    calibrated,
                    labels=np.arange(
                        NUM_CLASSES
                    ),
                ),
            }
        )

    candidate_df = pd.DataFrame(
        candidate_rows
    )

    best_row = candidate_df.sort_values(
        [
            "nll",
            "temperature",
        ],
        ascending=[
            True,
            True,
        ],
    ).iloc[0]

    best_temperature = float(
        best_row["temperature"]
    )

    TEMPERATURES[
        model_name
    ] = best_temperature

    CALIBRATED_VAL[
        model_name
    ] = temperature_scale(
        VAL_PROBABILITIES[
            model_name
        ],
        best_temperature,
    )

    CALIBRATED_TEST[
        model_name
    ] = temperature_scale(
        TEST_PROBABILITIES[
            model_name
        ],
        best_temperature,
    )

    temperature_rows.append(
        {
            "model": model_name,
            "temperature": best_temperature,
            "validation_nll": float(
                best_row["nll"]
            ),
        }
    )

TEMPERATURE_DF = pd.DataFrame(
    temperature_rows
)

display(TEMPERATURE_DF)

TEMPERATURE_DF.to_csv(
    OUTPUT_DIR
    / "temperature_scaling.csv",
    index=False,
)
````

````python
# ============================================================
# CELL 06: CONSTRAINED ALL-MODEL WEIGHTED SEARCH
# ============================================================

RANDOM_WEIGHT_CANDIDATES = 50000
WEIGHT_CHUNK_SIZE = 128
ANCHOR_WEIGHT_MIN = 0.35
ANCHOR_WEIGHT_MAX = 0.95
MIN_FOLD_WINS = 4


def bacc_many(
    prediction_matrix,
    y_true,
    indices=None,
):
    prediction_matrix = np.asarray(
        prediction_matrix
    )
    y_true = np.asarray(
        y_true,
        dtype=np.int32,
    )

    if indices is not None:
        prediction_matrix = (
            prediction_matrix[
                :,
                indices,
            ]
        )
        y_true = y_true[indices]

    recalls = []

    for class_id in np.unique(
        y_true
    ):
        mask = y_true == class_id

        recalls.append(
            np.mean(
                prediction_matrix[
                    :,
                    mask,
                ]
                == class_id,
                axis=1,
            )
        )

    return np.mean(
        np.stack(
            recalls,
            axis=1,
        ),
        axis=1,
    )


def generate_weight_candidates():
    n_models = len(MODEL_NAMES)
    anchor_index = MODEL_NAMES.index(
        ANCHOR_NAME
    )

    candidates = []

    # Deterministic pairwise blends.
    for auxiliary_index in range(
        n_models
    ):
        if auxiliary_index == anchor_index:
            continue

        for auxiliary_weight in np.arange(
            0.05,
            0.66,
            0.05,
        ):
            weights = np.zeros(
                n_models,
                dtype=np.float64,
            )
            weights[
                anchor_index
            ] = 1.0 - auxiliary_weight
            weights[
                auxiliary_index
            ] = auxiliary_weight

            if (
                ANCHOR_WEIGHT_MIN
                <= weights[anchor_index]
                <= ANCHOR_WEIGHT_MAX
            ):
                candidates.append(
                    weights
                )

    # Random constrained Dirichlet blends.
    anchor_weights = RNG.uniform(
        ANCHOR_WEIGHT_MIN,
        ANCHOR_WEIGHT_MAX,
        size=RANDOM_WEIGHT_CANDIDATES,
    )

    auxiliary_indices = [
        index
        for index in range(
            n_models
        )
        if index != anchor_index
    ]

    rest = RNG.dirichlet(
        np.full(
            len(auxiliary_indices),
            0.70,
        ),
        size=RANDOM_WEIGHT_CANDIDATES,
    )

    random_weights = np.zeros(
        (
            RANDOM_WEIGHT_CANDIDATES,
            n_models,
        ),
        dtype=np.float64,
    )

    random_weights[
        :,
        anchor_index,
    ] = anchor_weights

    random_weights[
        :,
        auxiliary_indices,
    ] = (
        rest
        * (
            1.0
            - anchor_weights[
                :,
                None,
            ]
        )
    )

    candidates.extend(
        random_weights
    )

    candidates = np.asarray(
        candidates,
        dtype=np.float32,
    )

    active_models = np.sum(
        candidates > 1e-4,
        axis=1,
    )
    candidates = candidates[
        active_models >= 2
    ]

    rounded = np.round(
        candidates,
        6,
    )
    _, unique_indices = np.unique(
        rounded,
        axis=0,
        return_index=True,
    )

    return candidates[
        np.sort(unique_indices)
    ]


VAL_STACK = np.stack(
    [
        CALIBRATED_VAL[
            model_name
        ]
        for model_name in MODEL_NAMES
    ],
    axis=0,
)

TEST_STACK = np.stack(
    [
        CALIBRATED_TEST[
            model_name
        ]
        for model_name in MODEL_NAMES
    ],
    axis=0,
)

WEIGHT_CANDIDATES = (
    generate_weight_candidates()
)

search_parts = []

for start in range(
    0,
    len(WEIGHT_CANDIDATES),
    WEIGHT_CHUNK_SIZE,
):
    stop = min(
        start + WEIGHT_CHUNK_SIZE,
        len(WEIGHT_CANDIDATES),
    )

    weights = WEIGHT_CANDIDATES[
        start:stop
    ]

    probabilities = np.einsum(
        "bm,mnc->bnc",
        weights,
        VAL_STACK,
        optimize=True,
    )

    predictions = np.argmax(
        probabilities,
        axis=2,
    )

    full_bacc = bacc_many(
        predictions,
        Y_VAL,
    )

    fold_scores = np.stack(
        [
            bacc_many(
                predictions,
                Y_VAL,
                fold_indices,
            )
            for _, fold_indices
            in FOLDS
        ],
        axis=1,
    )

    part = pd.DataFrame(
        {
            "candidate_index": np.arange(
                start,
                stop,
            ),
            "cv_mean_bacc": fold_scores.mean(
                axis=1
            ),
            "cv_std_bacc": fold_scores.std(
                axis=1
            ),
            "full_val_bacc": full_bacc,
            "full_val_gain_over_no_tta": (
                full_bacc
                - ANCHOR_FULL_BACC
            ),
            "fold_wins_over_no_tta": np.sum(
                fold_scores
                > (
                    ANCHOR_FOLD_BACC[
                        None,
                        :,
                    ]
                    + 1e-12
                ),
                axis=1,
            ),
            "anchor_weight": weights[
                :,
                MODEL_NAMES.index(
                    ANCHOR_NAME
                ),
            ],
        }
    )

    for model_index, model_name in enumerate(
        MODEL_NAMES
    ):
        part[
            f"weight__{model_name}"
        ] = weights[
            :,
            model_index,
        ]

    search_parts.append(part)

WEIGHT_SEARCH_DF = pd.concat(
    search_parts,
    ignore_index=True,
).sort_values(
    [
        "cv_mean_bacc",
        "fold_wins_over_no_tta",
        "full_val_bacc",
        "cv_std_bacc",
    ],
    ascending=[
        False,
        False,
        False,
        True,
    ],
).reset_index(drop=True)

ROBUST_WEIGHT_DF = WEIGHT_SEARCH_DF[
    (
        WEIGHT_SEARCH_DF[
            "fold_wins_over_no_tta"
        ]
        >= MIN_FOLD_WINS
    )
    & (
        WEIGHT_SEARCH_DF[
            "full_val_gain_over_no_tta"
        ]
        > 0.0
    )
]

if len(ROBUST_WEIGHT_DF) > 0:
    WEIGHT_SELECTION_PASSED = True
    BEST_WEIGHT_ROW = ROBUST_WEIGHT_DF.iloc[0]
else:
    WEIGHT_SELECTION_PASSED = False
    BEST_WEIGHT_ROW = WEIGHT_SEARCH_DF.iloc[0]

BEST_WEIGHTS = np.asarray(
    [
        BEST_WEIGHT_ROW[
            f"weight__{model_name}"
        ]
        for model_name in MODEL_NAMES
    ],
    dtype=np.float32,
)
BEST_WEIGHTS /= BEST_WEIGHTS.sum()

WEIGHTED_VAL_PROB = (
    normalize_probabilities(
        np.tensordot(
            BEST_WEIGHTS,
            VAL_STACK,
            axes=(0, 0),
        )
    )
)

WEIGHTED_TEST_PROB = (
    normalize_probabilities(
        np.tensordot(
            BEST_WEIGHTS,
            TEST_STACK,
            axes=(0, 0),
        )
    )
)

WEIGHT_SEARCH_DF.to_csv(
    OUTPUT_DIR
    / "weighted_search_validation.csv",
    index=False,
)

print(
    "Weight candidates:",
    len(WEIGHT_SEARCH_DF),
)
print(
    "Robust candidates:",
    len(ROBUST_WEIGHT_DF),
)
print(
    "Robust selection passed:",
    WEIGHT_SELECTION_PASSED,
)

display(
    pd.DataFrame(
        {
            "model": MODEL_NAMES,
            "weight": BEST_WEIGHTS,
        }
    )
)
````

````python
# ============================================================
# CELL 07: VALIDATION-FOLD STACKING AND NO-TTA BLEND
# ============================================================

STACK_C_VALUES = [
    0.01,
    0.03,
    0.10,
    0.30,
    1.00,
    3.00,
    10.00,
]

STACK_ALPHA_VALUES = np.round(
    np.arange(
        0.05,
        0.66,
        0.05,
    ),
    2,
)


def entropy(probabilities):
    probabilities = np.clip(
        probabilities,
        1e-12,
        1.0,
    )

    return -np.sum(
        probabilities
        * np.log(probabilities),
        axis=1,
        keepdims=True,
    )


def build_meta_features(
    source_dictionary,
):
    blocks = []

    for model_name in MODEL_NAMES:
        probabilities = (
            source_dictionary[
                model_name
            ]
        )

        blocks.extend(
            [
                probabilities,
                np.log(
                    np.clip(
                        probabilities,
                        1e-12,
                        1.0,
                    )
                ),
                np.max(
                    probabilities,
                    axis=1,
                    keepdims=True,
                ),
                entropy(
                    probabilities
                ),
            ]
        )

    return np.concatenate(
        blocks,
        axis=1,
    ).astype(np.float32)


def make_meta_model(c_value):
    return Pipeline(
        [
            (
                "scaler",
                StandardScaler(),
            ),
            (
                "classifier",
                LogisticRegression(
                    C=float(c_value),
                    class_weight="balanced",
                    solver="lbfgs",
                    max_iter=5000,
                    random_state=RANDOM_STATE,
                ),
            ),
        ]
    )


def align_probability(
    raw_probability,
    classes_seen,
):
    aligned = np.zeros(
        (
            raw_probability.shape[0],
            NUM_CLASSES,
        ),
        dtype=np.float32,
    )

    for source_column, class_id in enumerate(
        classes_seen
    ):
        aligned[
            :,
            int(class_id),
        ] = raw_probability[
            :,
            source_column,
        ]

    return normalize_probabilities(
        aligned
    )


META_X_VAL = build_meta_features(
    CALIBRATED_VAL
)
META_X_TEST = build_meta_features(
    CALIBRATED_TEST
)

stack_rows = []
stack_oof_probabilities = {}

for c_value in STACK_C_VALUES:
    oof_probability = np.zeros(
        (
            len(Y_VAL),
            NUM_CLASSES,
        ),
        dtype=np.float32,
    )

    for train_indices, fold_indices in FOLDS:
        model = make_meta_model(
            c_value
        )

        model.fit(
            META_X_VAL[
                train_indices
            ],
            Y_VAL[
                train_indices
            ],
        )

        raw_probability = model.predict_proba(
            META_X_VAL[
                fold_indices
            ]
        )

        classes_seen = (
            model.named_steps[
                "classifier"
            ].classes_
        )

        oof_probability[
            fold_indices
        ] = align_probability(
            raw_probability,
            classes_seen,
        )

    stack_oof_probabilities[
        c_value
    ] = oof_probability

    for alpha in STACK_ALPHA_VALUES:
        blended_probability = (
            normalize_probabilities(
                (
                    1.0 - alpha
                )
                * CALIBRATED_VAL[
                    ANCHOR_NAME
                ]
                + alpha
                * oof_probability
            )
        )

        blended_prediction = np.argmax(
            blended_probability,
            axis=1,
        )

        fold_bacc = np.asarray(
            [
                balanced_accuracy_score(
                    Y_VAL[
                        fold_indices
                    ],
                    blended_prediction[
                        fold_indices
                    ],
                )
                for _, fold_indices
                in FOLDS
            ]
        )

        full_bacc = (
            balanced_accuracy_score(
                Y_VAL,
                blended_prediction,
            )
        )

        stack_rows.append(
            {
                "C": float(c_value),
                "alpha": float(alpha),
                "cv_mean_bacc": float(
                    fold_bacc.mean()
                ),
                "cv_std_bacc": float(
                    fold_bacc.std()
                ),
                "full_val_bacc": float(
                    full_bacc
                ),
                "full_val_gain_over_no_tta": float(
                    full_bacc
                    - ANCHOR_FULL_BACC
                ),
                "fold_wins_over_no_tta": int(
                    np.sum(
                        fold_bacc
                        > (
                            ANCHOR_FOLD_BACC
                            + 1e-12
                        )
                    )
                ),
                "macro_f1": float(
                    f1_score(
                        Y_VAL,
                        blended_prediction,
                        average="macro",
                        zero_division=0,
                    )
                ),
            }
        )

STACK_SEARCH_DF = pd.DataFrame(
    stack_rows
).sort_values(
    [
        "cv_mean_bacc",
        "fold_wins_over_no_tta",
        "full_val_bacc",
        "macro_f1",
        "cv_std_bacc",
    ],
    ascending=[
        False,
        False,
        False,
        False,
        True,
    ],
).reset_index(drop=True)

ROBUST_STACK_DF = STACK_SEARCH_DF[
    (
        STACK_SEARCH_DF[
            "fold_wins_over_no_tta"
        ]
        >= MIN_FOLD_WINS
    )
    & (
        STACK_SEARCH_DF[
            "full_val_gain_over_no_tta"
        ]
        > 0.0
    )
]

if len(ROBUST_STACK_DF) > 0:
    STACK_SELECTION_PASSED = True
    BEST_STACK_ROW = ROBUST_STACK_DF.iloc[0]
else:
    STACK_SELECTION_PASSED = False
    BEST_STACK_ROW = STACK_SEARCH_DF.iloc[0]

BEST_C = float(
    BEST_STACK_ROW["C"]
)
BEST_ALPHA = float(
    BEST_STACK_ROW["alpha"]
)

STACK_OOF_PROB = (
    stack_oof_probabilities[
        BEST_C
    ]
)

STACK_BLEND_VAL_PROB = (
    normalize_probabilities(
        (
            1.0 - BEST_ALPHA
        )
        * CALIBRATED_VAL[
            ANCHOR_NAME
        ]
        + BEST_ALPHA
        * STACK_OOF_PROB
    )
)

FINAL_META_MODEL = make_meta_model(
    BEST_C
)
FINAL_META_MODEL.fit(
    META_X_VAL,
    Y_VAL,
)

raw_test_probability = (
    FINAL_META_MODEL.predict_proba(
        META_X_TEST
    )
)

STACK_TEST_PROB = align_probability(
    raw_test_probability,
    FINAL_META_MODEL.named_steps[
        "classifier"
    ].classes_,
)

STACK_BLEND_TEST_PROB = (
    normalize_probabilities(
        (
            1.0 - BEST_ALPHA
        )
        * CALIBRATED_TEST[
            ANCHOR_NAME
        ]
        + BEST_ALPHA
        * STACK_TEST_PROB
    )
)

STACK_SEARCH_DF.to_csv(
    OUTPUT_DIR
    / "stacking_validation_search.csv",
    index=False,
)

joblib.dump(
    FINAL_META_MODEL,
    OUTPUT_DIR
    / "stacking_meta_model.joblib",
)

print(
    "Robust stacking candidates:",
    len(ROBUST_STACK_DF),
)
print(
    "Robust selection passed:",
    STACK_SELECTION_PASSED,
)
print(
    "Selected C:",
    BEST_C,
)
print(
    "Selected alpha:",
    BEST_ALPHA,
)
display(
    BEST_STACK_ROW.to_frame().T
)
````

````python
# ============================================================
# CELL 08: OPTIONAL VALIDATION-ONLY CLASS-BIAS CORRECTION
# ============================================================

BIAS_VALUES = [
    -0.15,
    -0.10,
    -0.05,
    0.00,
    0.05,
    0.10,
    0.15,
]


def apply_class_bias(
    probabilities,
    class_bias,
):
    logits = np.log(
        np.clip(
            normalize_probabilities(
                probabilities
            ),
            1e-12,
            1.0,
        )
    )

    logits += np.asarray(
        class_bias,
        dtype=np.float64,
    )[None, :]

    logits -= logits.max(
        axis=1,
        keepdims=True,
    )

    return normalize_probabilities(
        np.exp(logits)
    )


def candidate_statistics(
    probabilities,
):
    predictions = np.argmax(
        probabilities,
        axis=1,
    )

    fold_bacc = np.asarray(
        [
            balanced_accuracy_score(
                Y_VAL[
                    fold_indices
                ],
                predictions[
                    fold_indices
                ],
            )
            for _, fold_indices
            in FOLDS
        ]
    )

    return {
        "cv_mean_bacc": float(
            fold_bacc.mean()
        ),
        "cv_std_bacc": float(
            fold_bacc.std()
        ),
        "full_val_bacc": float(
            balanced_accuracy_score(
                Y_VAL,
                predictions,
            )
        ),
        "fold_wins_over_no_tta": int(
            np.sum(
                fold_bacc
                > (
                    ANCHOR_FOLD_BACC
                    + 1e-12
                )
            )
        ),
        "macro_f1": float(
            f1_score(
                Y_VAL,
                predictions,
                average="macro",
                zero_division=0,
            )
        ),
    }


CLASS_BIAS = np.zeros(
    NUM_CLASSES,
    dtype=np.float32,
)

for _ in range(2):
    for class_id in range(
        NUM_CLASSES
    ):
        best_choice = None

        for candidate_bias in BIAS_VALUES:
            trial_bias = CLASS_BIAS.copy()
            trial_bias[
                class_id
            ] = candidate_bias

            trial_probability = (
                apply_class_bias(
                    STACK_BLEND_VAL_PROB,
                    trial_bias,
                )
            )

            statistics = (
                candidate_statistics(
                    trial_probability
                )
            )

            score = (
                statistics[
                    "cv_mean_bacc"
                ]
                - 1e-4
                * np.sum(
                    np.abs(
                        trial_bias
                    )
                ),
                statistics[
                    "fold_wins_over_no_tta"
                ],
                statistics[
                    "full_val_bacc"
                ],
                statistics[
                    "macro_f1"
                ],
                -abs(candidate_bias),
            )

            if (
                best_choice is None
                or score
                > best_choice[0]
            ):
                best_choice = (
                    score,
                    candidate_bias,
                )

        CLASS_BIAS[
            class_id
        ] = best_choice[1]

STACK_BIAS_VAL_PROB = (
    apply_class_bias(
        STACK_BLEND_VAL_PROB,
        CLASS_BIAS,
    )
)

STACK_BIAS_TEST_PROB = (
    apply_class_bias(
        STACK_BLEND_TEST_PROB,
        CLASS_BIAS,
    )
)

print(
    "Selected class bias:",
    CLASS_BIAS.tolist(),
)
print(
    "No-bias statistics:",
    candidate_statistics(
        STACK_BLEND_VAL_PROB
    ),
)
print(
    "Class-bias statistics:",
    candidate_statistics(
        STACK_BIAS_VAL_PROB
    ),
)
````

````python
# ============================================================
# CELL 09: FREEZE THE VALIDATION WINNER, THEN EVALUATE TEST
# ============================================================

VALIDATION_CANDIDATES = OrderedDict(
    [
        (
            "OptimizedWeightedEnsemble",
            (
                WEIGHTED_VAL_PROB,
                WEIGHTED_TEST_PROB,
            ),
        ),
        (
            "NoTTA_StackingBlend",
            (
                STACK_BLEND_VAL_PROB,
                STACK_BLEND_TEST_PROB,
            ),
        ),
        (
            "NoTTA_StackingBlend_ClassBias",
            (
                STACK_BIAS_VAL_PROB,
                STACK_BIAS_TEST_PROB,
            ),
        ),
    ]
)

candidate_rows = []

for method_name, (
    validation_probability,
    test_probability,
) in VALIDATION_CANDIDATES.items():
    statistics = candidate_statistics(
        validation_probability
    )

    candidate_rows.append(
        {
            "method": method_name,
            **statistics,
            "full_val_gain_over_no_tta": (
                statistics[
                    "full_val_bacc"
                ]
                - ANCHOR_FULL_BACC
            ),
        }
    )

CANDIDATE_DF = pd.DataFrame(
    candidate_rows
).sort_values(
    [
        "cv_mean_bacc",
        "fold_wins_over_no_tta",
        "full_val_bacc",
        "macro_f1",
        "cv_std_bacc",
    ],
    ascending=[
        False,
        False,
        False,
        False,
        True,
    ],
).reset_index(drop=True)

ROBUST_FINAL_DF = CANDIDATE_DF[
    (
        CANDIDATE_DF[
            "fold_wins_over_no_tta"
        ]
        >= MIN_FOLD_WINS
    )
    & (
        CANDIDATE_DF[
            "full_val_gain_over_no_tta"
        ]
        > 0.0
    )
]

if len(ROBUST_FINAL_DF) > 0:
    FINAL_SELECTION_PASSED = True
    FINAL_SELECTION_ROW = (
        ROBUST_FINAL_DF.iloc[0]
    )
else:
    FINAL_SELECTION_PASSED = False
    FINAL_SELECTION_ROW = (
        CANDIDATE_DF.iloc[0]
    )

FINAL_METHOD = str(
    FINAL_SELECTION_ROW["method"]
)

FINAL_VAL_PROB = (
    VALIDATION_CANDIDATES[
        FINAL_METHOD
    ][0]
)

FINAL_TEST_PROB = (
    VALIDATION_CANDIDATES[
        FINAL_METHOD
    ][1]
)

print("=" * 88)
print("VALIDATION-SELECTED FINAL ENSEMBLE")
print("=" * 88)
display(CANDIDATE_DF)
print(
    "\nFinal method:",
    FINAL_METHOD,
)
print(
    "Robust selection passed:",
    FINAL_SELECTION_PASSED,
)

# Test is evaluated only after the method above is frozen.
comparison_rows = []
result_objects = {}

for model_name in MODEL_NAMES:
    result = evaluate_probabilities(
        Y_TEST,
        TEST_PROBABILITIES[
            model_name
        ],
        model_name,
        "InternalTest",
        MODEL_CATEGORY.get(
            model_name,
            "Other",
        ),
    )

    result_objects[
        model_name
    ] = result

    comparison_rows.append(
        {
            key: value
            for key, value
            in result.items()
            if key not in {
                "y_pred",
                "confusion_matrix",
                "classification_report",
            }
        }
    )

for method_name, (
    validation_probability,
    test_probability,
) in VALIDATION_CANDIDATES.items():
    category = (
        "ProposedFinal"
        if method_name == FINAL_METHOD
        else "Ensemble"
    )

    result = evaluate_probabilities(
        Y_TEST,
        test_probability,
        method_name,
        "InternalTest",
        category,
    )

    result_objects[
        method_name
    ] = result

    comparison_rows.append(
        {
            key: value
            for key, value
            in result.items()
            if key not in {
                "y_pred",
                "confusion_matrix",
                "classification_report",
            }
        }
    )

# Reuse previous fixed weighted-ensemble predictions.
OLD_ENSEMBLE_VAL = (
    PIPELINE_ROOT
    / "simclr_dominant_weighted_ensemble_v2"
    / "WeightedEnsemble_val_predictions.csv"
)

OLD_ENSEMBLE_TEST = (
    PIPELINE_ROOT
    / "simclr_dominant_weighted_ensemble_v2"
    / "WeightedEnsemble_test_predictions.csv"
)

if (
    OLD_ENSEMBLE_VAL.exists()
    and OLD_ENSEMBLE_TEST.exists()
):
    old_y_test, old_test_prob = (
        load_prediction_csv(
            OLD_ENSEMBLE_TEST
        )
    )

    if np.array_equal(
        old_y_test,
        Y_TEST,
    ):
        old_result = (
            evaluate_probabilities(
                Y_TEST,
                old_test_prob,
                (
                    "PreviousFixed"
                    "WeightedEnsemble"
                ),
                "InternalTest",
                "Ensemble",
            )
        )

        result_objects[
            "PreviousFixedWeightedEnsemble"
        ] = old_result

        comparison_rows.append(
            {
                key: value
                for key, value
                in old_result.items()
                if key not in {
                    "y_pred",
                    "confusion_matrix",
                    "classification_report",
                }
            }
        )

COMPARISON_DF = pd.DataFrame(
    comparison_rows
).sort_values(
    [
        "bacc",
        "macro_f1",
        "accuracy",
    ],
    ascending=[
        False,
        False,
        False,
    ],
).reset_index(drop=True)

COMPARISON_DF[
    "rank_by_test_bacc"
] = np.arange(
    1,
    len(COMPARISON_DF) + 1,
)

display(COMPARISON_DF)

final_result = result_objects[
    FINAL_METHOD
]
simclr_result = result_objects[
    "SimCLR_NoTTA"
]
rotation_result = result_objects[
    "Rotation"
]

single_bacc = [
    result["bacc"]
    for result in result_objects.values()
    if result.get(
        "category"
    ) == "Single"
]

fusion_bacc = [
    result["bacc"]
    for result in result_objects.values()
    if result.get(
        "category"
    ) == "Fusion"
]

BEAT_FLAGS = {
    "final_method": FINAL_METHOD,
    "final_test_bacc": float(
        final_result["bacc"]
    ),
    "simclr_no_tta_test_bacc": float(
        simclr_result["bacc"]
    ),
    "rotation_test_bacc": float(
        rotation_result["bacc"]
    ),
    "beats_simclr_no_tta": bool(
        final_result["bacc"]
        > simclr_result["bacc"]
    ),
    "beats_rotation": bool(
        final_result["bacc"]
        > rotation_result["bacc"]
    ),
    "beats_best_single_model": bool(
        final_result["bacc"]
        > max(single_bacc)
    ),
    "beats_every_fusion_model": bool(
        final_result["bacc"]
        > max(fusion_bacc)
    ),
}

print("\n" + "=" * 88)
print("EXPLICIT FINAL ANSWERS")
print("=" * 88)
print(
    json.dumps(
        BEAT_FLAGS,
        indent=2,
    )
)
````

````python
# ============================================================
# EXTERNAL CELL E02: REMAP PREPARED ISIC/PAD TABLES TO INPUT IMAGES
# ============================================================

PREPARED_EXTERNAL_DIR = (
    PIPELINE_ROOT
    / "external_datasets_prepared"
)

ISIC_PREPARED_CSV = (
    PREPARED_EXTERNAL_DIR
    / "isic2019_official_test_prepared.csv"
)

PAD_PREPARED_CSV = (
    PREPARED_EXTERNAL_DIR
    / "pad_ufes_20_external_prepared.csv"
)

for required_path in [
    ISIC_PREPARED_CSV,
    PAD_PREPARED_CSV,
]:
    if not required_path.exists():
        raise FileNotFoundError(
            "The previous notebook's prepared external table "
            f"was not found: {required_path}"
        )

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp",
}

# Optional manual hints. Leave as None for automatic matching.
# Example:
# ISIC_PATH_HINT = "isic-2019-test"
# PAD_PATH_HINT = "pad-ufes-20"
ISIC_PATH_HINT = None
PAD_PATH_HINT = None


def normalized_image_key(value):
    value = str(value).strip()
    return Path(value).stem.lower()


def build_input_image_index():
    image_index = {}

    for path in Path("/kaggle/input").rglob("*"):
        if (
            path.is_file()
            and path.suffix.lower() in IMAGE_EXTENSIONS
        ):
            key = path.stem.lower()
            image_index.setdefault(
                key,
                [],
            ).append(path)

    return image_index


def choose_image_candidate(
    candidates,
    dataset_hint,
    manual_hint=None,
):
    if not candidates:
        return None

    ranked = []

    for candidate in candidates:
        text = str(candidate).lower()
        score = 0

        if manual_hint:
            if str(manual_hint).lower() in text:
                score += 100

        if dataset_hint == "isic":
            if "isic" in text:
                score += 30
            if "test" in text:
                score += 10
            if "pad" in text:
                score -= 50

        if dataset_hint == "pad":
            if "pad" in text:
                score += 30
            if "ufes" in text:
                score += 20
            if "isic" in text:
                score -= 50

        ranked.append(
            (
                -score,
                len(candidate.parts),
                str(candidate),
                candidate,
            )
        )

    ranked.sort()
    return ranked[0][-1]


def remap_prepared_dataframe(
    prepared_csv,
    dataset_hint,
    manual_hint=None,
):
    dataframe = pd.read_csv(
        prepared_csv
    )

    if "image_id" not in dataframe.columns:
        if "image_path" in dataframe.columns:
            dataframe["image_id"] = dataframe[
                "image_path"
            ].map(
                lambda value: Path(
                    str(value)
                ).stem
            )
        else:
            raise ValueError(
                f"{prepared_csv} has no image_id or image_path."
            )

    mapped_paths = []

    for image_id in dataframe["image_id"]:
        key = normalized_image_key(
            image_id
        )
        candidates = INPUT_IMAGE_INDEX.get(
            key,
            [],
        )
        selected = choose_image_candidate(
            candidates,
            dataset_hint=dataset_hint,
            manual_hint=manual_hint,
        )
        mapped_paths.append(
            str(selected)
            if selected is not None
            else None
        )

    dataframe = dataframe.copy()
    dataframe["image_path_current"] = mapped_paths
    dataframe["image_found"] = dataframe[
        "image_path_current"
    ].notna()

    missing_count = int(
        (~dataframe["image_found"]).sum()
    )

    print(
        f"{dataset_hint.upper()} prepared rows:",
        len(dataframe),
    )
    print(
        f"{dataset_hint.upper()} images mapped:",
        int(dataframe["image_found"].sum()),
    )
    print(
        f"{dataset_hint.upper()} images missing:",
        missing_count,
    )

    if missing_count > 0:
        display(
            dataframe.loc[
                ~dataframe["image_found"],
                ["image_id"],
            ].head(20)
        )

    dataframe = dataframe.loc[
        dataframe["image_found"]
    ].copy().reset_index(drop=True)

    if "label" not in dataframe.columns:
        raise ValueError(
            f"{prepared_csv} has no label column."
        )

    dataframe["label"] = dataframe[
        "label"
    ].fillna(-1).astype(int)

    return dataframe


print("Indexing attached input images...")
INPUT_IMAGE_INDEX = build_input_image_index()
print(
    "Unique image IDs indexed:",
    len(INPUT_IMAGE_INDEX),
)

ISIC_EXTERNAL_DF = remap_prepared_dataframe(
    ISIC_PREPARED_CSV,
    dataset_hint="isic",
    manual_hint=ISIC_PATH_HINT,
)

PAD_EXTERNAL_DF = remap_prepared_dataframe(
    PAD_PREPARED_CSV,
    dataset_hint="pad",
    manual_hint=PAD_PATH_HINT,
)

if len(ISIC_EXTERNAL_DF) == 0:
    raise RuntimeError(
        "No ISIC images were mapped. Set ISIC_PATH_HINT."
    )

if len(PAD_EXTERNAL_DF) == 0:
    raise RuntimeError(
        "No PAD images were mapped. Set PAD_PATH_HINT."
    )

print("\nISIC label distribution:")
print(
    ISIC_EXTERNAL_DF[
        "label"
    ].value_counts().sort_index()
)

print("\nPAD label distribution:")
print(
    PAD_EXTERNAL_DF[
        "label"
    ].value_counts().sort_index()
)
````

````python
# ============================================================
# EXTERNAL CELL E03: BUILD DETERMINISTIC IMAGE DATASETS
# ============================================================

def decode_external_image(
    image_path,
    label,
):
    image_bytes = tf.io.read_file(
        image_path
    )

    image = tf.io.decode_image(
        image_bytes,
        channels=3,
        expand_animations=False,
    )

    image.set_shape(
        [None, None, 3]
    )

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE],
        method="bilinear",
        antialias=True,
    )

    image = tf.cast(
        image,
        tf.float32,
    ) / 255.0

    return image, tf.cast(
        label,
        tf.int32,
    )


def build_external_dataset(
    dataframe,
    batch_size=EXTERNAL_BATCH_SIZE,
):
    dataset = tf.data.Dataset.from_tensor_slices(
        (
            dataframe[
                "image_path_current"
            ].astype(str).values,
            dataframe[
                "label"
            ].astype(np.int32).values,
        )
    )

    dataset = dataset.map(
        decode_external_image,
        num_parallel_calls=tf.data.AUTOTUNE,
        deterministic=True,
    )

    dataset = dataset.batch(
        batch_size,
        drop_remainder=False,
    )

    dataset = dataset.prefetch(
        tf.data.AUTOTUNE
    )

    return dataset


ISIC_EXTERNAL_DS = build_external_dataset(
    ISIC_EXTERNAL_DF
)

PAD_EXTERNAL_DS = build_external_dataset(
    PAD_EXTERNAL_DF
)

EXTERNAL_DATASETS_FOR_BOTH_STACKS = OrderedDict(
    [
        (
            "ISIC_2019_Official_Test",
            {
                "dataframe": ISIC_EXTERNAL_DF,
                "dataset": ISIC_EXTERNAL_DS,
            },
        ),
        (
            "PAD_UFES_20_External",
            {
                "dataframe": PAD_EXTERNAL_DF,
                "dataset": PAD_EXTERNAL_DS,
            },
        ),
    ]
)

for dataset_name, payload in (
    EXTERNAL_DATASETS_FOR_BOTH_STACKS.items()
):
    first_images, first_labels = next(
        iter(payload["dataset"])
    )
    print(
        dataset_name,
        "batch:",
        first_images.shape,
        first_labels.shape,
    )
````

````python
# ============================================================
# EXTERNAL CELL E04: INFERENCE-ONLY MODEL ARCHITECTURES
# ============================================================

SIMCLR_DROPOUT = 0.30
ROTATION_FT_DROPOUT = 0.30

FUSIONCONCAT_DROPOUT = 0.40
FUSIONCONCAT_HIDDEN_DIM_1 = 512
FUSIONCONCAT_HIDDEN_DIM_2 = 256
FUSIONCONCAT_WEIGHT_DECAY = 1e-4

FUSIONGATED_V2_DROPOUT = 0.50
FUSIONGATED_V2_WEIGHT_DECAY = 3e-4
FUSIONGATED_V2_BRANCH_DIM = 256
FUSIONGATED_V2_GATE_HIDDEN = 128
FUSIONGATED_V2_HEAD_HIDDEN = 128
FUSIONGATED_V2_TEMPERATURE = 1.5

BESTV3_BRANCH_DIM = 256
BESTV3_DROPOUT_1 = 0.50
BESTV3_DROPOUT_2 = 0.40
BESTV3_WEIGHT_DECAY = 4e-4
BESTV3_GATE_HIDDEN = 128
BESTV3_HEAD_HIDDEN_1 = 256
BESTV3_HEAD_HIDDEN_2 = 128
BESTV3_TEMPERATURE = 1.0

RESIDUAL_FUSION_BRANCH_DIM = 256
RESIDUAL_FUSION_HIDDEN_DIM = 256
RESIDUAL_FUSION_WEIGHT_DECAY = 3e-4
RESIDUAL_FUSION_DROPOUT = 0.40
RESIDUAL_FUSION_MAX_ALPHA = 0.35


def resnet_preprocess_tensor(images):
    images = tf.cast(
        images,
        tf.float32,
    ) * 255.0

    return tf.keras.applications.resnet50.preprocess_input(
        images
    )


def build_resnet50_encoder(
    name,
    training_flag=False,
):
    inputs = keras.Input(
        shape=(
            IMG_SIZE,
            IMG_SIZE,
            3,
        ),
        name=f"{name}_input",
    )

    backbone = ResNet50(
        include_top=False,
        weights=None,
        input_shape=(
            IMG_SIZE,
            IMG_SIZE,
            3,
        ),
        pooling="avg",
    )

    features = backbone(
        inputs,
        training=training_flag,
    )

    return keras.Model(
        inputs,
        features,
        name=name,
    )


def build_simclr_finetune_model():
    encoder = build_resnet50_encoder(
        "simclr_resnet50_encoder",
        training_flag=False,
    )

    inputs = keras.Input(
        shape=(
            IMG_SIZE,
            IMG_SIZE,
            3,
        ),
        name="finetune_img",
    )

    x = layers.Lambda(
        resnet_preprocess_tensor,
        name="resnet_preprocess",
    )(inputs)

    x = encoder(
        x,
        training=False,
    )

    x = layers.Dropout(
        SIMCLR_DROPOUT,
        name="finetune_dropout",
    )(x)

    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax",
        dtype="float32",
        name="finetune_classifier",
    )(x)

    model = keras.Model(
        inputs,
        outputs,
        name="simclr_finetune_model",
    )

    return model, encoder


def build_rotation_finetune_model():
    encoder = build_resnet50_encoder(
        "rotation_ft_resnet50_encoder",
        training_flag=True,
    )
    encoder.trainable = True

    inputs = keras.Input(
        shape=(
            IMG_SIZE,
            IMG_SIZE,
            3,
        ),
        name="imagenet_rotation_input",
    )

    x = layers.Lambda(
        resnet_preprocess_tensor,
        name="resnet50_preprocess",
    )(inputs)

    # This reproduces the saved architecture.
    x = encoder(
        x,
        training=True
        if encoder.trainable
        else False,
    )

    x = layers.Dropout(
        ROTATION_FT_DROPOUT,
        name="imagenet_rotation_dropout",
    )(x)

    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax",
        dtype="float32",
        name="imagenet_rotation_classifier",
    )(x)

    model = keras.Model(
        inputs,
        outputs,
        name="imagenet_rotation_finetune_model",
    )

    return model


def l2_normalize_features(
    features,
    epsilon=1e-12,
):
    features = np.asarray(
        features,
        dtype=np.float32,
    )

    denominator = np.linalg.norm(
        features,
        axis=1,
        keepdims=True,
    )

    return (
        features
        / np.clip(
            denominator,
            epsilon,
            None,
        )
    ).astype(np.float32)


def build_fusionconcat_model(
    input_dim,
):
    inputs = keras.Input(
        shape=(input_dim,),
        name="fusionconcat_input",
    )

    x = layers.BatchNormalization(
        name="bn_input"
    )(inputs)

    x = layers.Dense(
        FUSIONCONCAT_HIDDEN_DIM_1,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                FUSIONCONCAT_WEIGHT_DECAY
            )
        ),
        name="fusionconcat_dense_1",
    )(x)

    x = layers.BatchNormalization(
        name="bn_1"
    )(x)
    x = layers.Activation(
        "relu",
        name="relu_1",
    )(x)
    x = layers.Dropout(
        FUSIONCONCAT_DROPOUT,
        name="dropout_1",
    )(x)

    x = layers.Dense(
        FUSIONCONCAT_HIDDEN_DIM_2,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                FUSIONCONCAT_WEIGHT_DECAY
            )
        ),
        name="fusionconcat_dense_2",
    )(x)

    x = layers.BatchNormalization(
        name="bn_2"
    )(x)
    x = layers.Activation(
        "relu",
        name="relu_2",
    )(x)
    x = layers.Dropout(
        FUSIONCONCAT_DROPOUT,
        name="dropout_2",
    )(x)

    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax",
        dtype="float32",
        name="fusionconcat_classifier",
    )(x)

    return keras.Model(
        inputs,
        outputs,
        name="fusionconcat_model",
    )


def stable_softmax_temperature(
    values,
    temperature,
):
    return tf.nn.softmax(
        values / temperature,
        axis=-1,
    )


def build_fusiongated_v2_model(
    simclr_dim,
    rotation_dim,
):
    simclr_input = keras.Input(
        shape=(simclr_dim,),
        name="simclr_input",
    )
    rotation_input = keras.Input(
        shape=(rotation_dim,),
        name="rotation_input",
    )

    simclr_norm = layers.LayerNormalization(
        name="ln_simclr_in"
    )(simclr_input)

    rotation_norm = layers.LayerNormalization(
        name="ln_rotation_in"
    )(rotation_input)

    simclr_projection = layers.Dense(
        FUSIONGATED_V2_BRANCH_DIM,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                FUSIONGATED_V2_WEIGHT_DECAY
            )
        ),
        name="simclr_proj",
    )(simclr_norm)

    simclr_projection = layers.LayerNormalization(
        name="ln_simclr_proj"
    )(simclr_projection)
    simclr_projection = layers.Activation(
        "relu",
        name="relu_simclr_proj",
    )(simclr_projection)

    rotation_projection = layers.Dense(
        FUSIONGATED_V2_BRANCH_DIM,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                FUSIONGATED_V2_WEIGHT_DECAY
            )
        ),
        name="rotation_proj",
    )(rotation_norm)

    rotation_projection = layers.LayerNormalization(
        name="ln_rotation_proj"
    )(rotation_projection)
    rotation_projection = layers.Activation(
        "relu",
        name="relu_rotation_proj",
    )(rotation_projection)

    gate_input = layers.Concatenate(
        name="gate_input_concat"
    )(
        [
            simclr_projection,
            rotation_projection,
        ]
    )

    gate_hidden = layers.Dense(
        FUSIONGATED_V2_GATE_HIDDEN,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                FUSIONGATED_V2_WEIGHT_DECAY
            )
        ),
        name="gate_dense_1",
    )(gate_input)

    gate_hidden = layers.LayerNormalization(
        name="ln_gate_1"
    )(gate_hidden)
    gate_hidden = layers.Activation(
        "relu",
        name="relu_gate_1",
    )(gate_hidden)
    gate_hidden = layers.Dropout(
        FUSIONGATED_V2_DROPOUT,
        name="dropout_gate_1",
    )(gate_hidden)

    gate_logits = layers.Dense(
        2,
        activation=None,
        bias_initializer=(
            keras.initializers.Constant(
                [0.30, 0.00]
            )
        ),
        name="gate_logits",
    )(gate_hidden)

    gate_weights = layers.Lambda(
        lambda values: (
            stable_softmax_temperature(
                values,
                FUSIONGATED_V2_TEMPERATURE,
            )
        ),
        name="branch_gate_softmax",
    )(gate_logits)

    alpha = layers.Lambda(
        lambda values: values[:, 0:1],
        name="alpha_weight",
    )(gate_weights)

    beta = layers.Lambda(
        lambda values: values[:, 1:2],
        name="beta_weight",
    )(gate_weights)

    weighted_simclr = layers.Multiply(
        name="weighted_simclr"
    )(
        [
            simclr_projection,
            alpha,
        ]
    )

    weighted_rotation = layers.Multiply(
        name="weighted_rotation"
    )(
        [
            rotation_projection,
            beta,
        ]
    )

    weighted_sum = layers.Add(
        name="weighted_sum"
    )(
        [
            weighted_simclr,
            weighted_rotation,
        ]
    )

    fused = layers.Concatenate(
        name="residual_fused_concat"
    )(
        [
            weighted_sum,
            simclr_projection,
            rotation_projection,
        ]
    )

    x = layers.LayerNormalization(
        name="ln_fused"
    )(fused)

    x = layers.Dense(
        FUSIONGATED_V2_HEAD_HIDDEN,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                FUSIONGATED_V2_WEIGHT_DECAY
            )
        ),
        name="head_dense_1",
    )(x)

    x = layers.LayerNormalization(
        name="ln_head_1"
    )(x)
    x = layers.Activation(
        "relu",
        name="relu_head_1",
    )(x)
    x = layers.Dropout(
        FUSIONGATED_V2_DROPOUT,
        name="dropout_head_1",
    )(x)

    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax",
        dtype="float32",
        name="classifier",
    )(x)

    return keras.Model(
        [
            simclr_input,
            rotation_input,
        ],
        outputs,
        name="fusiongated_v2_model",
    )


def build_fusiongated_v3_model(
    simclr_dim,
    rotation_dim,
):
    simclr_input = keras.Input(
        shape=(simclr_dim,),
        name="simclr_input",
    )
    rotation_input = keras.Input(
        shape=(rotation_dim,),
        name="rotation_input",
    )

    simclr_norm = layers.LayerNormalization(
        name="ln_simclr_in"
    )(simclr_input)
    rotation_norm = layers.LayerNormalization(
        name="ln_rotation_in"
    )(rotation_input)

    simclr_projection = layers.Dense(
        BESTV3_BRANCH_DIM,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                BESTV3_WEIGHT_DECAY
            )
        ),
        name="simclr_proj",
    )(simclr_norm)
    simclr_projection = layers.LayerNormalization(
        name="ln_simclr_proj"
    )(simclr_projection)
    simclr_projection = layers.Activation(
        "relu",
        name="relu_simclr_proj",
    )(simclr_projection)

    rotation_projection = layers.Dense(
        BESTV3_BRANCH_DIM,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                BESTV3_WEIGHT_DECAY
            )
        ),
        name="rotation_proj",
    )(rotation_norm)
    rotation_projection = layers.LayerNormalization(
        name="ln_rotation_proj"
    )(rotation_projection)
    rotation_projection = layers.Activation(
        "relu",
        name="relu_rotation_proj",
    )(rotation_projection)

    gate_input = layers.Concatenate(
        name="gate_input_concat"
    )(
        [
            simclr_projection,
            rotation_projection,
        ]
    )

    gate_hidden = layers.Dense(
        BESTV3_GATE_HIDDEN,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                BESTV3_WEIGHT_DECAY
            )
        ),
        name="gate_dense_1",
    )(gate_input)
    gate_hidden = layers.LayerNormalization(
        name="ln_gate_1"
    )(gate_hidden)
    gate_hidden = layers.Activation(
        "relu",
        name="relu_gate_1",
    )(gate_hidden)
    gate_hidden = layers.Dropout(
        BESTV3_DROPOUT_1,
        name="dropout_gate_1",
    )(gate_hidden)

    gate_logits = layers.Dense(
        2,
        activation=None,
        bias_initializer=(
            keras.initializers.Constant(
                [0.35, 0.00]
            )
        ),
        name="gate_logits",
    )(gate_hidden)

    gate_weights = layers.Lambda(
        lambda values: (
            stable_softmax_temperature(
                values,
                BESTV3_TEMPERATURE,
            )
        ),
        name="branch_gate_softmax",
    )(gate_logits)

    alpha = layers.Lambda(
        lambda values: values[:, 0:1],
        name="alpha_weight",
    )(gate_weights)
    beta = layers.Lambda(
        lambda values: values[:, 1:2],
        name="beta_weight",
    )(gate_weights)

    weighted_simclr = layers.Multiply(
        name="weighted_simclr"
    )(
        [
            simclr_projection,
            alpha,
        ]
    )
    weighted_rotation = layers.Multiply(
        name="weighted_rotation"
    )(
        [
            rotation_projection,
            beta,
        ]
    )
    weighted_sum = layers.Add(
        name="weighted_sum"
    )(
        [
            weighted_simclr,
            weighted_rotation,
        ]
    )

    absolute_difference = layers.Lambda(
        lambda values: tf.abs(
            values[0] - values[1]
        ),
        name="abs_diff",
    )(
        [
            simclr_projection,
            rotation_projection,
        ]
    )

    elementwise_product = layers.Multiply(
        name="elem_prod"
    )(
        [
            simclr_projection,
            rotation_projection,
        ]
    )

    fused = layers.Concatenate(
        name="final_fused_concat"
    )(
        [
            weighted_sum,
            simclr_projection,
            rotation_projection,
            absolute_difference,
            elementwise_product,
        ]
    )

    x = layers.LayerNormalization(
        name="ln_fused"
    )(fused)

    x = layers.Dense(
        BESTV3_HEAD_HIDDEN_1,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                BESTV3_WEIGHT_DECAY
            )
        ),
        name="head_dense_1",
    )(x)
    x = layers.LayerNormalization(
        name="ln_head_1"
    )(x)
    x = layers.Activation(
        "relu",
        name="relu_head_1",
    )(x)
    x = layers.Dropout(
        BESTV3_DROPOUT_1,
        name="dropout_head_1",
    )(x)

    x = layers.Dense(
        BESTV3_HEAD_HIDDEN_2,
        activation=None,
        kernel_regularizer=(
            keras.regularizers.l2(
                BESTV3_WEIGHT_DECAY
            )
        ),
        name="head_dense_2",
    )(x)
    x = layers.LayerNormalization(
        name="ln_head_2"
    )(x)
    x = layers.Activation(
        "relu",
        name="relu_head_2",
    )(x)
    x = layers.Dropout(
        BESTV3_DROPOUT_2,
        name="dropout_head_2",
    )(x)

    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax",
        dtype="float32",
        name="classifier",
    )(x)

    return keras.Model(
        [
            simclr_input,
            rotation_input,
        ],
        outputs,
        name="fusiongated_best_v3_model",
    )


class BoundedResidualLogitFusion(
    layers.Layer
):
    def __init__(
        self,
        max_alpha=0.35,
        initial_alpha=0.08,
        **kwargs,
    ):
        super().__init__(**kwargs)
        self.max_alpha = float(
            max_alpha
        )
        ratio = np.clip(
            initial_alpha / max_alpha,
            1e-4,
            1.0 - 1e-4,
        )
        self.initial_logit = float(
            np.log(
                ratio
                / (1.0 - ratio)
            )
        )

    def build(
        self,
        input_shape,
    ):
        self.alpha_logit = (
            self.add_weight(
                name="alpha_logit",
                shape=(),
                initializer=(
                    keras.initializers.Constant(
                        self.initial_logit
                    )
                ),
                trainable=True,
            )
        )
        super().build(
            input_shape
        )

    def call(
        self,
        inputs,
    ):
        base_logits, correction_logits = inputs
        alpha = (
            self.max_alpha
            * tf.sigmoid(
                self.alpha_logit
            )
        )
        return (
            base_logits
            + alpha
            * correction_logits
        )

    def get_config(self):
        config = super().get_config()
        config.update(
            {
                "max_alpha": self.max_alpha,
            }
        )
        return config


def build_residual_fusion_model(
    simclr_dim,
    rotation_dim,
):
    simclr_input = keras.Input(
        shape=(simclr_dim,),
        name="simclr_feature_input",
    )
    rotation_input = keras.Input(
        shape=(rotation_dim,),
        name="rotation_feature_input",
    )
    base_logits_input = keras.Input(
        shape=(NUM_CLASSES,),
        name="base_simclr_logits",
    )

    simclr_projection = (
        layers.LayerNormalization(
            name="sim_ln_in"
        )(simclr_input)
    )
    simclr_projection = layers.Dense(
        RESIDUAL_FUSION_BRANCH_DIM,
        kernel_regularizer=(
            keras.regularizers.l2(
                RESIDUAL_FUSION_WEIGHT_DECAY
            )
        ),
        name="sim_projection",
    )(simclr_projection)
    simclr_projection = (
        layers.LayerNormalization(
            name="sim_ln_proj"
        )(simclr_projection)
    )
    simclr_projection = layers.Activation(
        "relu",
        name="sim_relu",
    )(simclr_projection)

    rotation_projection = (
        layers.LayerNormalization(
            name="rot_ln_in"
        )(rotation_input)
    )
    rotation_projection = layers.Dense(
        RESIDUAL_FUSION_BRANCH_DIM,
        kernel_regularizer=(
            keras.regularizers.l2(
                RESIDUAL_FUSION_WEIGHT_DECAY
            )
        ),
        name="rot_projection",
    )(rotation_projection)
    rotation_projection = (
        layers.LayerNormalization(
            name="rot_ln_proj"
        )(rotation_projection)
    )
    rotation_projection = layers.Activation(
        "relu",
        name="rot_relu",
    )(rotation_projection)

    absolute_difference = layers.Lambda(
        lambda values: tf.abs(
            values[0] - values[1]
        ),
        name="abs_difference",
    )(
        [
            simclr_projection,
            rotation_projection,
        ]
    )

    product = layers.Multiply(
        name="elementwise_product"
    )(
        [
            simclr_projection,
            rotation_projection,
        ]
    )

    fused = layers.Concatenate(
        name="residual_feature_concat"
    )(
        [
            simclr_projection,
            rotation_projection,
            absolute_difference,
            product,
        ]
    )
    fused = layers.LayerNormalization(
        name="fusion_ln"
    )(fused)
    fused = layers.Dense(
        RESIDUAL_FUSION_HIDDEN_DIM,
        kernel_regularizer=(
            keras.regularizers.l2(
                RESIDUAL_FUSION_WEIGHT_DECAY
            )
        ),
        name="correction_dense",
    )(fused)
    fused = layers.LayerNormalization(
        name="correction_ln"
    )(fused)
    fused = layers.Activation(
        "relu",
        name="correction_relu",
    )(fused)
    fused = layers.Dropout(
        RESIDUAL_FUSION_DROPOUT,
        name="correction_dropout",
    )(fused)
    correction_logits = layers.Dense(
        NUM_CLASSES,
        name="correction_logits",
    )(fused)

    final_logits = (
        BoundedResidualLogitFusion(
            max_alpha=(
                RESIDUAL_FUSION_MAX_ALPHA
            ),
            initial_alpha=0.08,
            name=(
                "bounded_residual_"
                "logit_fusion"
            ),
        )(
            [
                base_logits_input,
                correction_logits,
            ]
        )
    )

    output = layers.Softmax(
        dtype="float32",
        name="final_probabilities",
    )(final_logits)

    return keras.Model(
        [
            simclr_input,
            rotation_input,
            base_logits_input,
        ],
        output,
        name=(
            "simclr_preserving_"
            "residual_fusion"
        ),
    )
````

````python
# ============================================================
# EXTERNAL CELL E06: BASE-MODEL EXTERNAL INFERENCE WITH CACHING
# ============================================================

def apply_tta_transform(
    images,
    transform_name,
):
    if transform_name == "identity":
        return images
    if transform_name == "hflip":
        return tf.image.flip_left_right(
            images
        )
    if transform_name == "vflip":
        return tf.image.flip_up_down(
            images
        )
    if transform_name == "rot90":
        return tf.image.rot90(
            images,
            k=1,
        )
    if transform_name == "rot180":
        return tf.image.rot90(
            images,
            k=2,
        )
    if transform_name == "rot270":
        return tf.image.rot90(
            images,
            k=3,
        )

    raise ValueError(
        f"Unknown TTA transform: {transform_name}"
    )


def predict_simclr_seed_both(
    model,
    dataset,
    tta_transforms,
):
    no_tta_batches = []
    selected_tta_batches = []
    labels = []

    for images, batch_labels in dataset:
        identity_probability = np.asarray(
            model(
                images,
                training=False,
            ),
            dtype=np.float32,
        )

        view_probabilities = []

        for transform_name in tta_transforms:
            if transform_name == "identity":
                transformed_probability = (
                    identity_probability
                )
            else:
                transformed_images = (
                    apply_tta_transform(
                        images,
                        transform_name,
                    )
                )
                transformed_probability = (
                    np.asarray(
                        model(
                            transformed_images,
                            training=False,
                        ),
                        dtype=np.float32,
                    )
                )

            view_probabilities.append(
                transformed_probability
            )

        selected_probability = np.mean(
            np.stack(
                view_probabilities,
                axis=0,
            ),
            axis=0,
        )

        no_tta_batches.append(
            identity_probability
        )
        selected_tta_batches.append(
            selected_probability
        )
        labels.append(
            np.asarray(
                batch_labels,
                dtype=np.int32,
            )
        )

    return (
        normalize_probabilities(
            np.concatenate(
                no_tta_batches,
                axis=0,
            )
        ),
        normalize_probabilities(
            np.concatenate(
                selected_tta_batches,
                axis=0,
            )
        ),
        np.concatenate(
            labels,
            axis=0,
        ),
    )


def extract_encoder_features(
    encoder,
    dataset,
):
    feature_batches = []
    labels = []

    for images, batch_labels in dataset:
        preprocessed = (
            resnet_preprocess_tensor(
                images
            )
        )

        features = encoder(
            preprocessed,
            training=False,
        )

        feature_batches.append(
            np.asarray(
                features,
                dtype=np.float32,
            )
        )

        labels.append(
            np.asarray(
                batch_labels,
                dtype=np.int32,
            )
        )

    return (
        l2_normalize_features(
            np.concatenate(
                feature_batches,
                axis=0,
            )
        ),
        np.concatenate(
            labels,
            axis=0,
        ),
    )


def predict_direct_model(
    model,
    dataset,
):
    probability_batches = []
    labels = []

    for images, batch_labels in dataset:
        probability_batches.append(
            np.asarray(
                model(
                    images,
                    training=False,
                ),
                dtype=np.float32,
            )
        )
        labels.append(
            np.asarray(
                batch_labels,
                dtype=np.int32,
            )
        )

    return (
        normalize_probabilities(
            np.concatenate(
                probability_batches,
                axis=0,
            )
        ),
        np.concatenate(
            labels,
            axis=0,
        ),
    )


def generate_external_base_probabilities(
    dataset_name,
    dataset,
    expected_labels,
):
    cache_path = (
        EXTERNAL_CACHE_DIR
        / f"{dataset_name}_all_base_probabilities.npz"
    )

    if cache_path.exists():
        with np.load(
            cache_path
        ) as saved:
            cached_names = [
                str(value)
                for value in saved[
                    "model_names"
                ].tolist()
            ]

            if cached_names == MODEL_NAMES:
                cached_labels = np.asarray(
                    saved["labels"],
                    dtype=np.int32,
                )

                if np.array_equal(
                    cached_labels,
                    expected_labels,
                ):
                    print(
                        "Using cached external base probabilities:",
                        cache_path,
                    )

                    return OrderedDict(
                        [
                            (
                                model_name,
                                normalize_probabilities(
                                    saved[
                                        f"prob__{model_name}"
                                    ]
                                ),
                            )
                            for model_name
                            in MODEL_NAMES
                        ]
                    )

    print(
        "Generating external base probabilities for",
        dataset_name,
    )

    # --------------------------------------------------------
    # 1. Multi-seed SimCLR: no-TTA and selected TTA.
    # --------------------------------------------------------
    no_tta_seed_probabilities = []
    selected_tta_seed_probabilities = []
    representative_probability = None

    for seed, weight_path in (
        SIMCLR_SEED_PATHS.items()
    ):
        print(
            f"SimCLR external inference, seed {seed}"
        )

        tf.keras.backend.clear_session()
        gc.collect()

        model, _ = (
            build_simclr_finetune_model()
        )
        model.load_weights(
            str(weight_path)
        )

        (
            seed_no_tta,
            seed_selected_tta,
            labels_check,
        ) = predict_simclr_seed_both(
            model,
            dataset,
            SELECTED_TTA_TRANSFORMS,
        )

        if not np.array_equal(
            labels_check,
            expected_labels,
        ):
            raise ValueError(
                "External label order changed during "
                f"SimCLR seed {seed} inference."
            )

        no_tta_seed_probabilities.append(
            seed_no_tta
        )
        selected_tta_seed_probabilities.append(
            seed_selected_tta
        )

        if (
            seed
            == SIMCLR_REPRESENTATIVE_SEED
        ):
            representative_probability = (
                seed_no_tta
            )

        del model
        gc.collect()

    simclr_no_tta_probability = (
        normalize_probabilities(
            np.mean(
                np.stack(
                    no_tta_seed_probabilities,
                    axis=0,
                ),
                axis=0,
            )
        )
    )

    simclr_selected_tta_probability = (
        normalize_probabilities(
            np.mean(
                np.stack(
                    selected_tta_seed_probabilities,
                    axis=0,
                ),
                axis=0,
            )
        )
    )

    if representative_probability is None:
        raise RuntimeError(
            "The representative SimCLR seed probability "
            "was not generated."
        )

    # --------------------------------------------------------
    # 2. Extract representative SimCLR and Rotation features.
    # --------------------------------------------------------
    tf.keras.backend.clear_session()
    gc.collect()

    representative_model, representative_encoder = (
        build_simclr_finetune_model()
    )
    representative_model.load_weights(
        str(
            SIMCLR_SEED_PATHS[
                SIMCLR_REPRESENTATIVE_SEED
            ]
        )
    )

    (
        simclr_features,
        feature_labels,
    ) = extract_encoder_features(
        representative_encoder,
        dataset,
    )

    del representative_model
    del representative_encoder
    gc.collect()

    if not np.array_equal(
        feature_labels,
        expected_labels,
    ):
        raise ValueError(
            "SimCLR external feature label order changed."
        )

    tf.keras.backend.clear_session()
    gc.collect()

    rotation_encoder = build_resnet50_encoder(
        "rotation_resnet50_encoder",
        training_flag=False,
    )
    rotation_encoder.load_weights(
        str(
            ROTATION_SSL_ENCODER_PATH
        )
    )
    rotation_encoder.trainable = False

    (
        rotation_features,
        rotation_feature_labels,
    ) = extract_encoder_features(
        rotation_encoder,
        dataset,
    )

    del rotation_encoder
    gc.collect()

    if not np.array_equal(
        rotation_feature_labels,
        expected_labels,
    ):
        raise ValueError(
            "Rotation external feature label order changed."
        )

    concat_features = np.concatenate(
        [
            simclr_features,
            rotation_features,
        ],
        axis=1,
    ).astype(np.float32)

    # --------------------------------------------------------
    # 3. Standalone Rotation.
    # --------------------------------------------------------
    tf.keras.backend.clear_session()
    gc.collect()

    rotation_model = (
        build_rotation_finetune_model()
    )
    rotation_model.load_weights(
        str(ROTATION_FT_PATH)
    )

    (
        rotation_probability,
        rotation_labels,
    ) = predict_direct_model(
        rotation_model,
        dataset,
    )

    del rotation_model
    gc.collect()

    if not np.array_equal(
        rotation_labels,
        expected_labels,
    ):
        raise ValueError(
            "Standalone Rotation external label order changed."
        )

    # --------------------------------------------------------
    # 4. FusionConcat.
    # --------------------------------------------------------
    tf.keras.backend.clear_session()
    gc.collect()

    fusionconcat_model = (
        build_fusionconcat_model(
            concat_features.shape[1]
        )
    )
    fusionconcat_model.load_weights(
        str(FUSIONCONCAT_PATH)
    )

    fusionconcat_probability = (
        normalize_probabilities(
            fusionconcat_model.predict(
                concat_features,
                batch_size=256,
                verbose=0,
            )
        )
    )

    del fusionconcat_model
    gc.collect()

    # --------------------------------------------------------
    # 5. FusionGated V2.
    # --------------------------------------------------------
    tf.keras.backend.clear_session()
    gc.collect()

    fusiongated_v2_model = (
        build_fusiongated_v2_model(
            simclr_features.shape[1],
            rotation_features.shape[1],
        )
    )
    fusiongated_v2_model.load_weights(
        str(FUSIONGATED_V2_PATH)
    )

    fusiongated_v2_probability = (
        normalize_probabilities(
            fusiongated_v2_model.predict(
                [
                    simclr_features,
                    rotation_features,
                ],
                batch_size=256,
                verbose=0,
            )
        )
    )

    del fusiongated_v2_model
    gc.collect()

    # --------------------------------------------------------
    # 6. FusionGated V3.
    # --------------------------------------------------------
    tf.keras.backend.clear_session()
    gc.collect()

    fusiongated_v3_model = (
        build_fusiongated_v3_model(
            simclr_features.shape[1],
            rotation_features.shape[1],
        )
    )
    fusiongated_v3_model.load_weights(
        str(FUSIONGATED_V3_PATH)
    )

    fusiongated_v3_probability = (
        normalize_probabilities(
            fusiongated_v3_model.predict(
                [
                    simclr_features,
                    rotation_features,
                ],
                batch_size=256,
                verbose=0,
            )
        )
    )

    del fusiongated_v3_model
    gc.collect()

    # --------------------------------------------------------
    # 7. Residual Fusion multi-seed.
    # --------------------------------------------------------
    representative_logits = np.log(
        np.clip(
            representative_probability,
            1e-7,
            1.0,
        )
    ).astype(np.float32)

    residual_probabilities = []

    for seed, weight_path in (
        RESIDUAL_SEED_PATHS.items()
    ):
        print(
            f"Residual Fusion external inference, seed {seed}"
        )

        tf.keras.backend.clear_session()
        gc.collect()

        residual_model = (
            build_residual_fusion_model(
                simclr_features.shape[1],
                rotation_features.shape[1],
            )
        )
        residual_model.load_weights(
            str(weight_path)
        )

        residual_probabilities.append(
            normalize_probabilities(
                residual_model.predict(
                    [
                        simclr_features,
                        rotation_features,
                        representative_logits,
                    ],
                    batch_size=256,
                    verbose=0,
                )
            )
        )

        del residual_model
        gc.collect()

    residual_probability = (
        normalize_probabilities(
            np.mean(
                np.stack(
                    residual_probabilities,
                    axis=0,
                ),
                axis=0,
            )
        )
    )

    source_probabilities = OrderedDict(
        [
            (
                "SimCLR_NoTTA",
                simclr_no_tta_probability,
            ),
            (
                "SimCLR_SelectedTTA",
                simclr_selected_tta_probability,
            ),
            (
                "Rotation",
                rotation_probability,
            ),
            (
                "FusionConcat",
                fusionconcat_probability,
            ),
            (
                "FusionGatedV2",
                fusiongated_v2_probability,
            ),
            (
                "FusionGatedV3",
                fusiongated_v3_probability,
            ),
            (
                "ResidualFusion",
                residual_probability,
            ),
        ]
    )

    if list(
        source_probabilities.keys()
    ) != MODEL_NAMES:
        raise ValueError(
            "External source ordering does not match "
            "the frozen stacking model."
        )

    save_payload = {
        "model_names": np.asarray(
            MODEL_NAMES
        ),
        "labels": expected_labels,
    }

    for model_name, probability in (
        source_probabilities.items()
    ):
        save_payload[
            f"prob__{model_name}"
        ] = probability.astype(
            np.float32
        )

    np.savez_compressed(
        cache_path,
        **save_payload,
    )

    print(
        "Saved external base-probability cache:",
        cache_path,
    )

    return source_probabilities
````

````python
# ============================================================
# EXTERNAL CELL E07: APPLY BOTH FROZEN STACKING VARIANTS
# ============================================================

def apply_both_frozen_stacking_models(
    raw_source_probabilities,
):
    calibrated_sources = OrderedDict()

    for model_name in MODEL_NAMES:
        calibrated_sources[
            model_name
        ] = temperature_scale(
            raw_source_probabilities[
                model_name
            ],
            TEMPERATURES[
                model_name
            ],
        )

    meta_features = build_meta_features(
        calibrated_sources
    )

    raw_stacking_probability = (
        FINAL_META_MODEL.predict_proba(
            meta_features
        )
    )

    stacking_probability = align_probability(
        raw_stacking_probability,
        FINAL_META_MODEL.named_steps[
            "classifier"
        ].classes_,
    )

    no_bias_probability = (
        normalize_probabilities(
            (
                1.0 - BEST_ALPHA
            )
            * calibrated_sources[
                "SimCLR_NoTTA"
            ]
            + BEST_ALPHA
            * stacking_probability
        )
    )

    logits = np.log(
        np.clip(
            no_bias_probability,
            1e-12,
            1.0,
        )
    )

    logits = (
        logits
        + np.asarray(
            CLASS_BIAS,
            dtype=np.float64,
        )[None, :]
    )

    logits = (
        logits
        - logits.max(
            axis=1,
            keepdims=True,
        )
    )

    class_bias_probability = (
        normalize_probabilities(
            np.exp(logits)
        )
    )

    return {
        NO_BIAS_METHOD: no_bias_probability,
        CLASS_BIAS_METHOD: (
            class_bias_probability
        ),
    }
````

````python
# ============================================================
# EXTERNAL CELL E08: EXTERNAL METRICS AND OUTPUT HELPERS
# ============================================================

def external_macro_specificity(
    confusion,
):
    total = confusion.sum()
    specificities = []

    for class_id in range(
        confusion.shape[0]
    ):
        true_positive = confusion[
            class_id,
            class_id,
        ]
        false_negative = (
            confusion[
                class_id,
                :,
            ].sum()
            - true_positive
        )
        false_positive = (
            confusion[
                :,
                class_id,
            ].sum()
            - true_positive
        )
        true_negative = (
            total
            - true_positive
            - false_negative
            - false_positive
        )

        denominator = (
            true_negative
            + false_positive
        )

        specificities.append(
            true_negative / denominator
            if denominator > 0
            else 0.0
        )

    return float(
        np.mean(
            specificities
        )
    )


def evaluate_external_probability(
    y_true,
    probability,
    dataset_name,
    method_name,
    class_names=CLASS_NAMES,
):
    y_true = np.asarray(
        y_true,
        dtype=np.int32,
    )

    probability = (
        normalize_probabilities(
            probability
        )
    )

    labeled_mask = y_true >= 0

    if np.sum(
        labeled_mask
    ) == 0:
        return None

    y_labeled = y_true[
        labeled_mask
    ]

    probability_labeled = probability[
        labeled_mask
    ]

    prediction = np.argmax(
        probability_labeled,
        axis=1,
    )

    class_ids = np.arange(
        probability_labeled.shape[1]
    )

    confusion = confusion_matrix(
        y_labeled,
        prediction,
        labels=class_ids,
    )

    y_binary = label_binarize(
        y_labeled,
        classes=class_ids,
    )

    try:
        auc = float(
            roc_auc_score(
                y_binary,
                probability_labeled,
                average="macro",
                multi_class="ovr",
            )
        )
    except Exception:
        auc = float("nan")

    metrics = {
        "dataset": dataset_name,
        "method": method_name,
        "n_total": int(
            len(y_true)
        ),
        "n_labeled": int(
            np.sum(
                labeled_mask
            )
        ),
        "accuracy": float(
            accuracy_score(
                y_labeled,
                prediction,
            )
        ),
        "bacc": float(
            balanced_accuracy_score(
                y_labeled,
                prediction,
            )
        ),
        "macro_precision": float(
            precision_score(
                y_labeled,
                prediction,
                average="macro",
                zero_division=0,
            )
        ),
        "macro_recall": float(
            recall_score(
                y_labeled,
                prediction,
                average="macro",
                zero_division=0,
            )
        ),
        "macro_f1": float(
            f1_score(
                y_labeled,
                prediction,
                average="macro",
                zero_division=0,
            )
        ),
        "weighted_f1": float(
            f1_score(
                y_labeled,
                prediction,
                average="weighted",
                zero_division=0,
            )
        ),
        "macro_specificity": (
            external_macro_specificity(
                confusion
            )
        ),
        "auc_macro_ovr": auc,
    }

    report = classification_report(
        y_labeled,
        prediction,
        labels=class_ids,
        target_names=class_names,
        output_dict=True,
        zero_division=0,
    )

    return {
        "metrics": metrics,
        "confusion": confusion,
        "report": report,
        "prediction_labeled": prediction,
        "labeled_mask": labeled_mask,
    }


def save_external_method_outputs(
    dataset_name,
    dataframe,
    method_name,
    probability,
    result,
):
    method_dir = (
        EXTERNAL_OUTPUT_DIR
        / dataset_name
        / method_name
    )

    method_dir.mkdir(
        parents=True,
        exist_ok=True,
    )

    probability = (
        normalize_probabilities(
            probability
        )
    )

    prediction = np.argmax(
        probability,
        axis=1,
    )

    output_df = dataframe.copy()
    output_df["method"] = method_name
    output_df["pred_label"] = prediction
    output_df["pred_label_name"] = [
        CLASS_NAMES[
            class_id
        ]
        for class_id in prediction
    ]
    output_df["pred_confidence"] = np.max(
        probability,
        axis=1,
    )

    for class_id, class_name in enumerate(
        CLASS_NAMES
    ):
        output_df[
            f"prob_{class_id}_{class_name}"
        ] = probability[
            :,
            class_id,
        ]

    output_df.to_csv(
        method_dir
        / "predictions.csv",
        index=False,
    )

    if result is not None:
        pd.DataFrame(
            result[
                "confusion"
            ],
            index=[
                f"true_{name}"
                for name in CLASS_NAMES
            ],
            columns=[
                f"pred_{name}"
                for name in CLASS_NAMES
            ],
        ).to_csv(
            method_dir
            / "confusion_matrix.csv"
        )

        pd.DataFrame(
            result[
                "report"
            ]
        ).transpose().to_csv(
            method_dir
            / "classification_report.csv"
        )

        with open(
            method_dir
            / "metrics.json",
            "w",
        ) as file:
            json.dump(
                result[
                    "metrics"
                ],
                file,
                indent=2,
            )
````

# PAD-UFES-20 Fine-Tuned / Domain-Adapted Ensemble Extension

This section creates PAD-specific six-class versions of both final stacking
models without retraining the SSL encoders or feature-fusion networks.

## Domain-adaptation protocol

- Reuse the seven PAD base-model probability arrays already generated above.
- Reuse the original PAD 70/15/15 split: 1,608 training, 345 validation,
  and 345 test samples.
- Restrict predictions to the six PAD/ISIC common classes:
  `MEL, NV, BCC, AK, BKL, SCC`.
- Calibrate every base model using PAD training labels only.
- Train a class-balanced six-class logistic stacking adapter on PAD training.
- Select `C` and the no-TTA residual-blend strength using PAD validation only.
- Select optional six-class logit biases using PAD validation only.
- Evaluate the two frozen domain-adapted variants once on PAD test.

The two reported adapted models are:

1. **PAD-DA NoTTA Stacking Blend**
2. **PAD-DA NoTTA Stacking Blend + Class Bias**

This is a PAD fine-tuned/domain-adapted experiment, not a zero-shot result.

````python
# ============================================================
# PAD-DA CELL D02: REUSE THE ORIGINAL 70/15/15 PAD SPLIT
# ============================================================

def pad_da_image_key(value):
    return Path(
        str(value)
    ).stem.lower().strip()


pad_da_df = PAD_EXTERNAL_DF.copy().reset_index(
    drop=True
)

if "image_id" in pad_da_df.columns:
    pad_da_df["pad_da_key"] = pad_da_df[
        "image_id"
    ].map(
        pad_da_image_key
    )
elif "image_path_current" in pad_da_df.columns:
    pad_da_df["pad_da_key"] = pad_da_df[
        "image_path_current"
    ].map(
        pad_da_image_key
    )
else:
    raise ValueError(
        "PAD dataframe has no image_id or image_path_current."
    )

pad_da_df["label"] = pad_da_df[
    "label"
].astype(np.int32)

pad_da_df = pad_da_df.loc[
    pad_da_df["label"].isin(
        PAD_DA_CLASS_IDS_8
    )
].copy().reset_index(drop=True)

PAD_DA_ORIGINAL_TO_6 = {
    original_id: local_id
    for local_id, original_id
    in enumerate(
        PAD_DA_CLASS_IDS_8
    )
}

pad_da_df["label_6"] = pad_da_df[
    "label"
].map(
    PAD_DA_ORIGINAL_TO_6
).astype(np.int32)

# The old notebook saved the exact split used for the
# previously reported PAD fine-tuned FusionConcat result.
PAD_DA_PREVIOUS_DIR = (
    Path(PIPELINE_ROOT)
    / "pad_ufes_20_six_class_finetuning"
)

PAD_DA_PREVIOUS_SPLIT_CSV = (
    PAD_DA_PREVIOUS_DIR
    / "pad_six_class_split.csv"
)

PAD_DA_USED_PREVIOUS_SPLIT = False

if PAD_DA_PREVIOUS_SPLIT_CSV.exists():
    previous_split_df = pd.read_csv(
        PAD_DA_PREVIOUS_SPLIT_CSV
    )

    if "image_id" in previous_split_df.columns:
        previous_split_df[
            "pad_da_key"
        ] = previous_split_df[
            "image_id"
        ].map(
            pad_da_image_key
        )
    elif "image_path" in previous_split_df.columns:
        previous_split_df[
            "pad_da_key"
        ] = previous_split_df[
            "image_path"
        ].map(
            pad_da_image_key
        )
    elif "image_path_current" in previous_split_df.columns:
        previous_split_df[
            "pad_da_key"
        ] = previous_split_df[
            "image_path_current"
        ].map(
            pad_da_image_key
        )
    else:
        raise ValueError(
            "The saved PAD split has no usable image identifier."
        )

    if "split" not in previous_split_df.columns:
        raise ValueError(
            "The saved PAD split has no split column."
        )

    split_map_df = previous_split_df[
        [
            "pad_da_key",
            "split",
        ]
    ].drop_duplicates(
        subset=["pad_da_key"]
    )

    pad_da_df = pad_da_df.merge(
        split_map_df,
        on="pad_da_key",
        how="left",
        validate="one_to_one",
    )

    missing_split_count = int(
        pad_da_df["split"].isna().sum()
    )

    if missing_split_count == 0:
        PAD_DA_USED_PREVIOUS_SPLIT = True
    else:
        print(
            "Saved split could not map all images; "
            "recreating the deterministic split. "
            "Unmapped:",
            missing_split_count,
        )
        pad_da_df = pad_da_df.drop(
            columns=["split"]
        )

if not PAD_DA_USED_PREVIOUS_SPLIT:
    pad_da_train_df, pad_da_temp_df = (
        train_test_split(
            pad_da_df,
            test_size=0.30,
            stratify=pad_da_df[
                "label_6"
            ],
            random_state=PAD_DA_SEED,
        )
    )

    pad_da_val_df, pad_da_test_df = (
        train_test_split(
            pad_da_temp_df,
            test_size=0.50,
            stratify=pad_da_temp_df[
                "label_6"
            ],
            random_state=PAD_DA_SEED,
        )
    )

    pad_da_df["split"] = ""

    pad_da_df.loc[
        pad_da_train_df.index,
        "split",
    ] = "train"

    pad_da_df.loc[
        pad_da_val_df.index,
        "split",
    ] = "val"

    pad_da_df.loc[
        pad_da_test_df.index,
        "split",
    ] = "test"

valid_split_names = {
    "train",
    "val",
    "test",
}

if set(
    pad_da_df["split"].unique()
) != valid_split_names:
    raise ValueError(
        "PAD split must contain exactly train, val, and test."
    )

PAD_DA_TRAIN_INDEX = np.flatnonzero(
    pad_da_df["split"].values == "train"
)
PAD_DA_VAL_INDEX = np.flatnonzero(
    pad_da_df["split"].values == "val"
)
PAD_DA_TEST_INDEX = np.flatnonzero(
    pad_da_df["split"].values == "test"
)

PAD_DA_Y_ALL = pad_da_df[
    "label_6"
].to_numpy(
    dtype=np.int32
)

PAD_DA_Y_TRAIN = PAD_DA_Y_ALL[
    PAD_DA_TRAIN_INDEX
]
PAD_DA_Y_VAL = PAD_DA_Y_ALL[
    PAD_DA_VAL_INDEX
]
PAD_DA_Y_TEST = PAD_DA_Y_ALL[
    PAD_DA_TEST_INDEX
]

split_summary = (
    pad_da_df.groupby(
        [
            "split",
            "label_6",
        ]
    )
    .size()
    .unstack(
        fill_value=0
    )
)

print(
    "Reused previous exact PAD split:",
    PAD_DA_USED_PREVIOUS_SPLIT,
)
print(
    "Train / validation / test:",
    len(PAD_DA_TRAIN_INDEX),
    len(PAD_DA_VAL_INDEX),
    len(PAD_DA_TEST_INDEX),
)
display(split_summary)

if (
    len(PAD_DA_TRAIN_INDEX) != 1608
    or len(PAD_DA_VAL_INDEX) != 345
    or len(PAD_DA_TEST_INDEX) != 345
):
    print(
        "Warning: counts differ from the original "
        "1608/345/345 split."
    )

pad_da_df.to_csv(
    PAD_DA_OUTPUT_DIR
    / "pad_domain_adaptation_split.csv",
    index=False,
)
````

````python
# ============================================================
# PAD-DA CELL D03: SIX-CLASS PROBABILITIES AND TRAIN CALIBRATION
# ============================================================

def pad_da_restrict_to_six(
    probability_8,
):
    probability_8 = (
        normalize_probabilities(
            probability_8
        )
    )

    return normalize_probabilities(
        probability_8[
            :,
            PAD_DA_CLASS_IDS_8,
        ]
    )


PAD_DA_RAW_PROB_6 = OrderedDict(
    [
        (
            model_name,
            pad_da_restrict_to_six(
                PAD_DA_BASE_PROB_8[
                    model_name
                ]
            ),
        )
        for model_name
        in PAD_DA_MODEL_NAMES
    ]
)

PAD_DA_TEMPERATURE_GRID = [
    0.50,
    0.60,
    0.70,
    0.80,
    0.90,
    1.00,
    1.10,
    1.20,
    1.30,
    1.50,
    1.75,
    2.00,
    2.25,
    2.50,
]


def pad_da_temperature_scale(
    probability,
    temperature,
):
    probability = normalize_probabilities(
        probability
    )

    logits = np.log(
        np.clip(
            probability,
            1e-12,
            1.0,
        )
    )
    logits /= float(
        temperature
    )
    logits -= logits.max(
        axis=1,
        keepdims=True,
    )

    return normalize_probabilities(
        np.exp(logits)
    )


PAD_DA_TEMPERATURES = {}
PAD_DA_CALIBRATED_PROB_6 = OrderedDict()
pad_da_temperature_rows = []

for model_name, probability in (
    PAD_DA_RAW_PROB_6.items()
):
    candidate_rows = []

    for temperature in (
        PAD_DA_TEMPERATURE_GRID
    ):
        calibrated_train = (
            pad_da_temperature_scale(
                probability[
                    PAD_DA_TRAIN_INDEX
                ],
                temperature,
            )
        )

        candidate_rows.append(
            {
                "temperature": float(
                    temperature
                ),
                "train_nll": float(
                    log_loss(
                        PAD_DA_Y_TRAIN,
                        calibrated_train,
                        labels=np.arange(
                            PAD_DA_NUM_CLASSES
                        ),
                    )
                ),
            }
        )

    candidate_df = pd.DataFrame(
        candidate_rows
    ).sort_values(
        [
            "train_nll",
            "temperature",
        ],
        ascending=[
            True,
            True,
        ],
    )

    selected_row = candidate_df.iloc[
        0
    ]

    selected_temperature = float(
        selected_row["temperature"]
    )

    PAD_DA_TEMPERATURES[
        model_name
    ] = selected_temperature

    PAD_DA_CALIBRATED_PROB_6[
        model_name
    ] = pad_da_temperature_scale(
        probability,
        selected_temperature,
    )

    pad_da_temperature_rows.append(
        {
            "model": model_name,
            "selected_temperature": (
                selected_temperature
            ),
            "train_nll": float(
                selected_row[
                    "train_nll"
                ]
            ),
        }
    )

PAD_DA_TEMPERATURE_DF = pd.DataFrame(
    pad_da_temperature_rows
)

display(
    PAD_DA_TEMPERATURE_DF
)

PAD_DA_TEMPERATURE_DF.to_csv(
    PAD_DA_OUTPUT_DIR
    / "pad_train_temperature_scaling.csv",
    index=False,
)
````

````python
# ============================================================
# PAD-DA CELL D04: TRAIN/VALIDATION META-ADAPTER SEARCH
# ============================================================

def pad_da_entropy(
    probability,
):
    probability = np.clip(
        probability,
        1e-12,
        1.0,
    )

    return -np.sum(
        probability
        * np.log(
            probability
        ),
        axis=1,
        keepdims=True,
    )


def pad_da_build_meta_features(
    source_probabilities,
):
    blocks = []

    for model_name in (
        PAD_DA_MODEL_NAMES
    ):
        probability = (
            source_probabilities[
                model_name
            ]
        )

        blocks.extend(
            [
                probability,
                np.log(
                    np.clip(
                        probability,
                        1e-12,
                        1.0,
                    )
                ),
                np.max(
                    probability,
                    axis=1,
                    keepdims=True,
                ),
                pad_da_entropy(
                    probability
                ),
            ]
        )

    return np.concatenate(
        blocks,
        axis=1,
    ).astype(np.float32)


def pad_da_make_meta_model(
    c_value,
):
    return Pipeline(
        [
            (
                "scaler",
                StandardScaler(),
            ),
            (
                "classifier",
                LogisticRegression(
                    C=float(
                        c_value
                    ),
                    class_weight="balanced",
                    solver="lbfgs",
                    max_iter=5000,
                    random_state=PAD_DA_SEED,
                ),
            ),
        ]
    )


def pad_da_align_probability(
    raw_probability,
    classes_seen,
):
    aligned = np.zeros(
        (
            len(raw_probability),
            PAD_DA_NUM_CLASSES,
        ),
        dtype=np.float32,
    )

    for source_column, class_id in enumerate(
        classes_seen
    ):
        aligned[
            :,
            int(class_id),
        ] = raw_probability[
            :,
            source_column,
        ]

    return normalize_probabilities(
        aligned
    )


PAD_DA_META_X_ALL = (
    pad_da_build_meta_features(
        PAD_DA_CALIBRATED_PROB_6
    )
)

PAD_DA_C_VALUES = [
    0.001,
    0.003,
    0.01,
    0.03,
    0.10,
    0.30,
    1.00,
    3.00,
    10.00,
]

PAD_DA_ALPHA_VALUES = np.round(
    np.arange(
        0.00,
        1.01,
        0.05,
    ),
    2,
)

PAD_DA_ANCHOR_ALL = (
    PAD_DA_CALIBRATED_PROB_6[
        "SimCLR_NoTTA"
    ]
)

pad_da_search_rows = []
pad_da_models_by_c = {}

for c_value in PAD_DA_C_VALUES:
    model = pad_da_make_meta_model(
        c_value
    )

    model.fit(
        PAD_DA_META_X_ALL[
            PAD_DA_TRAIN_INDEX
        ],
        PAD_DA_Y_TRAIN,
    )

    raw_val_probability = (
        model.predict_proba(
            PAD_DA_META_X_ALL[
                PAD_DA_VAL_INDEX
            ]
        )
    )

    val_probability = (
        pad_da_align_probability(
            raw_val_probability,
            model.named_steps[
                "classifier"
            ].classes_,
        )
    )

    pad_da_models_by_c[
        float(c_value)
    ] = model

    for alpha in (
        PAD_DA_ALPHA_VALUES
    ):
        blended_val_probability = (
            normalize_probabilities(
                (
                    1.0 - alpha
                )
                * PAD_DA_ANCHOR_ALL[
                    PAD_DA_VAL_INDEX
                ]
                + alpha
                * val_probability
            )
        )

        val_prediction = np.argmax(
            blended_val_probability,
            axis=1,
        )

        pad_da_search_rows.append(
            {
                "C": float(
                    c_value
                ),
                "alpha_stack": float(
                    alpha
                ),
                "val_bacc": float(
                    balanced_accuracy_score(
                        PAD_DA_Y_VAL,
                        val_prediction,
                    )
                ),
                "val_macro_f1": float(
                    f1_score(
                        PAD_DA_Y_VAL,
                        val_prediction,
                        average="macro",
                        zero_division=0,
                    )
                ),
                "val_accuracy": float(
                    accuracy_score(
                        PAD_DA_Y_VAL,
                        val_prediction,
                    )
                ),
                "val_nll": float(
                    log_loss(
                        PAD_DA_Y_VAL,
                        blended_val_probability,
                        labels=np.arange(
                            PAD_DA_NUM_CLASSES
                        ),
                    )
                ),
            }
        )

PAD_DA_STACK_SEARCH_DF = pd.DataFrame(
    pad_da_search_rows
).sort_values(
    [
        "val_bacc",
        "val_macro_f1",
        "val_accuracy",
        "val_nll",
        "alpha_stack",
        "C",
    ],
    ascending=[
        False,
        False,
        False,
        True,
        True,
        True,
    ],
).reset_index(drop=True)

PAD_DA_BEST_STACK_ROW = (
    PAD_DA_STACK_SEARCH_DF.iloc[0]
)

PAD_DA_BEST_C = float(
    PAD_DA_BEST_STACK_ROW["C"]
)

PAD_DA_BEST_ALPHA = float(
    PAD_DA_BEST_STACK_ROW[
        "alpha_stack"
    ]
)

PAD_DA_META_MODEL = (
    pad_da_models_by_c[
        PAD_DA_BEST_C
    ]
)

raw_pad_da_val_stack = (
    PAD_DA_META_MODEL.predict_proba(
        PAD_DA_META_X_ALL[
            PAD_DA_VAL_INDEX
        ]
    )
)

PAD_DA_VAL_STACK_PROB = (
    pad_da_align_probability(
        raw_pad_da_val_stack,
        PAD_DA_META_MODEL.named_steps[
            "classifier"
        ].classes_,
    )
)

PAD_DA_NO_BIAS_VAL_PROB = (
    normalize_probabilities(
        (
            1.0 - PAD_DA_BEST_ALPHA
        )
        * PAD_DA_ANCHOR_ALL[
            PAD_DA_VAL_INDEX
        ]
        + PAD_DA_BEST_ALPHA
        * PAD_DA_VAL_STACK_PROB
    )
)

PAD_DA_STACK_SEARCH_DF.to_csv(
    PAD_DA_OUTPUT_DIR
    / "pad_meta_adapter_validation_search.csv",
    index=False,
)

joblib.dump(
    PAD_DA_META_MODEL,
    PAD_DA_OUTPUT_DIR
    / "pad_six_class_stacking_adapter.joblib",
)

print("=" * 92)
print("PAD META-ADAPTER VALIDATION SELECTION")
print("=" * 92)
print("Selected C:", PAD_DA_BEST_C)
print(
    "Selected stacking alpha:",
    PAD_DA_BEST_ALPHA,
)
display(
    PAD_DA_BEST_STACK_ROW.to_frame().T
)
````

````python
# ============================================================
# PAD-DA CELL D05: VALIDATION-ONLY SIX-CLASS BIAS SEARCH
# ============================================================

PAD_DA_BIAS_VALUES = [
    -0.30,
    -0.20,
    -0.15,
    -0.10,
    -0.05,
    0.00,
    0.05,
    0.10,
    0.15,
    0.20,
    0.30,
]


def pad_da_apply_bias(
    probability,
    bias,
):
    logits = np.log(
        np.clip(
            normalize_probabilities(
                probability
            ),
            1e-12,
            1.0,
        )
    )

    logits += np.asarray(
        bias,
        dtype=np.float64,
    )[None, :]

    logits -= logits.max(
        axis=1,
        keepdims=True,
    )

    return normalize_probabilities(
        np.exp(logits)
    )


def pad_da_validation_tuple(
    probability,
    bias=None,
):
    prediction = np.argmax(
        probability,
        axis=1,
    )

    regularization = (
        0.0
        if bias is None
        else 1e-4
        * float(
            np.sum(
                np.abs(
                    bias
                )
            )
        )
    )

    return (
        float(
            balanced_accuracy_score(
                PAD_DA_Y_VAL,
                prediction,
            )
        )
        - regularization,
        float(
            f1_score(
                PAD_DA_Y_VAL,
                prediction,
                average="macro",
                zero_division=0,
            )
        ),
        float(
            accuracy_score(
                PAD_DA_Y_VAL,
                prediction,
            )
        ),
        -float(
            log_loss(
                PAD_DA_Y_VAL,
                probability,
                labels=np.arange(
                    PAD_DA_NUM_CLASSES
                ),
            )
        ),
    )


PAD_DA_CLASS_BIAS = np.zeros(
    PAD_DA_NUM_CLASSES,
    dtype=np.float32,
)

for _ in range(3):
    for class_id in range(
        PAD_DA_NUM_CLASSES
    ):
        best_local = None

        for candidate_bias in (
            PAD_DA_BIAS_VALUES
        ):
            trial_bias = (
                PAD_DA_CLASS_BIAS.copy()
            )
            trial_bias[
                class_id
            ] = candidate_bias

            trial_probability = (
                pad_da_apply_bias(
                    PAD_DA_NO_BIAS_VAL_PROB,
                    trial_bias,
                )
            )

            score = (
                *pad_da_validation_tuple(
                    trial_probability,
                    trial_bias,
                ),
                -abs(
                    candidate_bias
                ),
            )

            if (
                best_local is None
                or score > best_local[0]
            ):
                best_local = (
                    score,
                    candidate_bias,
                )

        PAD_DA_CLASS_BIAS[
            class_id
        ] = best_local[1]

PAD_DA_CLASS_BIAS_VAL_PROB = (
    pad_da_apply_bias(
        PAD_DA_NO_BIAS_VAL_PROB,
        PAD_DA_CLASS_BIAS,
    )
)

PAD_DA_NO_BIAS_VAL_PRED = np.argmax(
    PAD_DA_NO_BIAS_VAL_PROB,
    axis=1,
)

PAD_DA_CLASS_BIAS_VAL_PRED = np.argmax(
    PAD_DA_CLASS_BIAS_VAL_PROB,
    axis=1,
)

PAD_DA_VAL_COMPARISON_DF = pd.DataFrame(
    [
        {
            "method": (
                "PAD_DA_NoTTA_StackingBlend"
            ),
            "val_bacc": float(
                balanced_accuracy_score(
                    PAD_DA_Y_VAL,
                    PAD_DA_NO_BIAS_VAL_PRED,
                )
            ),
            "val_macro_f1": float(
                f1_score(
                    PAD_DA_Y_VAL,
                    PAD_DA_NO_BIAS_VAL_PRED,
                    average="macro",
                    zero_division=0,
                )
            ),
            "val_accuracy": float(
                accuracy_score(
                    PAD_DA_Y_VAL,
                    PAD_DA_NO_BIAS_VAL_PRED,
                )
            ),
        },
        {
            "method": (
                "PAD_DA_NoTTA_StackingBlend_ClassBias"
            ),
            "val_bacc": float(
                balanced_accuracy_score(
                    PAD_DA_Y_VAL,
                    PAD_DA_CLASS_BIAS_VAL_PRED,
                )
            ),
            "val_macro_f1": float(
                f1_score(
                    PAD_DA_Y_VAL,
                    PAD_DA_CLASS_BIAS_VAL_PRED,
                    average="macro",
                    zero_division=0,
                )
            ),
            "val_accuracy": float(
                accuracy_score(
                    PAD_DA_Y_VAL,
                    PAD_DA_CLASS_BIAS_VAL_PRED,
                )
            ),
        },
    ]
).sort_values(
    [
        "val_bacc",
        "val_macro_f1",
        "val_accuracy",
    ],
    ascending=[
        False,
        False,
        False,
    ],
).reset_index(drop=True)

PAD_DA_SELECTED_METHOD = str(
    PAD_DA_VAL_COMPARISON_DF.iloc[
        0
    ]["method"]
)

print("=" * 92)
print("PAD DOMAIN-ADAPTED VARIANT VALIDATION COMPARISON")
print("=" * 92)
print(
    "Selected PAD class bias:",
    dict(
        zip(
            PAD_DA_CLASS_NAMES,
            [
                float(value)
                for value in PAD_DA_CLASS_BIAS
            ],
        )
    ),
)
display(
    PAD_DA_VAL_COMPARISON_DF
)
print(
    "PAD validation-selected adapted method:",
    PAD_DA_SELECTED_METHOD,
)
````

````python
# ============================================================
# PAD-DA CELL D06: FROZEN TEST EVALUATION OF BOTH ADAPTED MODELS
# ============================================================

raw_pad_da_test_stack = (
    PAD_DA_META_MODEL.predict_proba(
        PAD_DA_META_X_ALL[
            PAD_DA_TEST_INDEX
        ]
    )
)

PAD_DA_TEST_STACK_PROB = (
    pad_da_align_probability(
        raw_pad_da_test_stack,
        PAD_DA_META_MODEL.named_steps[
            "classifier"
        ].classes_,
    )
)

PAD_DA_NO_BIAS_TEST_PROB = (
    normalize_probabilities(
        (
            1.0 - PAD_DA_BEST_ALPHA
        )
        * PAD_DA_ANCHOR_ALL[
            PAD_DA_TEST_INDEX
        ]
        + PAD_DA_BEST_ALPHA
        * PAD_DA_TEST_STACK_PROB
    )
)

PAD_DA_CLASS_BIAS_TEST_PROB = (
    pad_da_apply_bias(
        PAD_DA_NO_BIAS_TEST_PROB,
        PAD_DA_CLASS_BIAS,
    )
)


def pad_da_evaluate(
    y_true,
    probability,
    method_name,
    result_type,
):
    probability = normalize_probabilities(
        probability
    )

    prediction = np.argmax(
        probability,
        axis=1,
    )

    confusion = confusion_matrix(
        y_true,
        prediction,
        labels=np.arange(
            PAD_DA_NUM_CLASSES
        ),
    )

    y_binary = label_binarize(
        y_true,
        classes=np.arange(
            PAD_DA_NUM_CLASSES
        ),
    )

    try:
        auc = float(
            roc_auc_score(
                y_binary,
                probability,
                average="macro",
                multi_class="ovr",
            )
        )
    except Exception:
        auc = float("nan")

    summary = {
        "method": method_name,
        "result_type": result_type,
        "n_samples": int(
            len(y_true)
        ),
        "accuracy": float(
            accuracy_score(
                y_true,
                prediction,
            )
        ),
        "bacc": float(
            balanced_accuracy_score(
                y_true,
                prediction,
            )
        ),
        "macro_precision": float(
            precision_score(
                y_true,
                prediction,
                average="macro",
                zero_division=0,
            )
        ),
        "macro_recall": float(
            recall_score(
                y_true,
                prediction,
                average="macro",
                zero_division=0,
            )
        ),
        "macro_f1": float(
            f1_score(
                y_true,
                prediction,
                average="macro",
                zero_division=0,
            )
        ),
        "weighted_f1": float(
            f1_score(
                y_true,
                prediction,
                average="weighted",
                zero_division=0,
            )
        ),
        "auc_macro_ovr": auc,
    }

    report = classification_report(
        y_true,
        prediction,
        labels=np.arange(
            PAD_DA_NUM_CLASSES
        ),
        target_names=(
            PAD_DA_CLASS_NAMES
        ),
        output_dict=True,
        zero_division=0,
    )

    return {
        "summary": summary,
        "prediction": prediction,
        "confusion": confusion,
        "report": report,
        "probability": probability,
    }


# ------------------------------------------------------------
# Same 345-image test-subset zero-shot references
# ------------------------------------------------------------
PAD_DA_ZERO_SHOT_NO_BIAS_ALL = (
    pad_da_restrict_to_six(
        PAD_DA_ZERO_SHOT_ENSEMBLES_8[
            "NoTTA_StackingBlend"
        ]
    )
)

PAD_DA_ZERO_SHOT_CLASS_BIAS_ALL = (
    pad_da_restrict_to_six(
        PAD_DA_ZERO_SHOT_ENSEMBLES_8[
            "NoTTA_StackingBlend_ClassBias"
        ]
    )
)

pad_da_result_objects = OrderedDict(
    [
        (
            "ZeroShot_NoTTA_StackingBlend",
            pad_da_evaluate(
                PAD_DA_Y_TEST,
                PAD_DA_ZERO_SHOT_NO_BIAS_ALL[
                    PAD_DA_TEST_INDEX
                ],
                "ZeroShot_NoTTA_StackingBlend",
                "Zero-shot on PAD adapted-test subset",
            ),
        ),
        (
            "ZeroShot_NoTTA_StackingBlend_ClassBias",
            pad_da_evaluate(
                PAD_DA_Y_TEST,
                PAD_DA_ZERO_SHOT_CLASS_BIAS_ALL[
                    PAD_DA_TEST_INDEX
                ],
                (
                    "ZeroShot_NoTTA_"
                    "StackingBlend_ClassBias"
                ),
                "Zero-shot on PAD adapted-test subset",
            ),
        ),
        (
            "PAD_DA_NoTTA_StackingBlend",
            pad_da_evaluate(
                PAD_DA_Y_TEST,
                PAD_DA_NO_BIAS_TEST_PROB,
                "PAD_DA_NoTTA_StackingBlend",
                "PAD fine-tuned/domain-adapted test",
            ),
        ),
        (
            "PAD_DA_NoTTA_StackingBlend_ClassBias",
            pad_da_evaluate(
                PAD_DA_Y_TEST,
                PAD_DA_CLASS_BIAS_TEST_PROB,
                (
                    "PAD_DA_NoTTA_"
                    "StackingBlend_ClassBias"
                ),
                "PAD fine-tuned/domain-adapted test",
            ),
        ),
    ]
)

PAD_DA_TEST_RESULTS_DF = pd.DataFrame(
    [
        result["summary"]
        for result in (
            pad_da_result_objects.values()
        )
    ]
)

# Add the previous FusionConcat PAD adapter as a reference
# when its saved metric file is available.
previous_pad_metrics_csv = (
    PAD_DA_PREVIOUS_DIR
    / "pad_six_class_test_metrics.csv"
)

if previous_pad_metrics_csv.exists():
    previous_metrics_df = pd.read_csv(
        previous_pad_metrics_csv
    )

    if len(previous_metrics_df) > 0:
        previous_row = (
            previous_metrics_df.iloc[
                0
            ].to_dict()
        )

        def pad_da_previous_metric(
            *names,
        ):
            for name in names:
                if name in previous_row:
                    try:
                        return float(
                            previous_row[name]
                        )
                    except Exception:
                        pass
            return np.nan

        PAD_DA_TEST_RESULTS_DF = pd.concat(
            [
                PAD_DA_TEST_RESULTS_DF,
                pd.DataFrame(
                    [
                        {
                            "method": (
                                "Previous_PAD_FineTuned_"
                                "FusionConcat_Adapter"
                            ),
                            "result_type": (
                                "Previous PAD fine-tuned/"
                                "domain-adapted test reference"
                            ),
                            "n_samples": int(
                                pad_da_previous_metric(
                                    "n_samples",
                                    "samples",
                                )
                                if not np.isnan(
                                    pad_da_previous_metric(
                                        "n_samples",
                                        "samples",
                                    )
                                )
                                else len(
                                    PAD_DA_Y_TEST
                                )
                            ),
                            "accuracy": (
                                pad_da_previous_metric(
                                    "accuracy"
                                )
                            ),
                            "bacc": (
                                pad_da_previous_metric(
                                    "bacc",
                                    "balanced_accuracy",
                                )
                            ),
                            "macro_precision": (
                                pad_da_previous_metric(
                                    "macro_precision"
                                )
                            ),
                            "macro_recall": (
                                pad_da_previous_metric(
                                    "macro_recall"
                                )
                            ),
                            "macro_f1": (
                                pad_da_previous_metric(
                                    "macro_f1"
                                )
                            ),
                            "weighted_f1": (
                                pad_da_previous_metric(
                                    "weighted_f1"
                                )
                            ),
                            "auc_macro_ovr": (
                                pad_da_previous_metric(
                                    "auc_macro_ovr",
                                    "auc",
                                )
                            ),
                        }
                    ]
                ),
            ],
            ignore_index=True,
        )

PAD_DA_TEST_RESULTS_DF = (
    PAD_DA_TEST_RESULTS_DF.sort_values(
        [
            "bacc",
            "macro_f1",
            "accuracy",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    ).reset_index(drop=True)
)

PAD_DA_TEST_RESULTS_DF[
    "rank_by_test_bacc"
] = np.arange(
    1,
    len(
        PAD_DA_TEST_RESULTS_DF
    ) + 1,
)

print("=" * 92)
print("PAD FINE-TUNED / DOMAIN-ADAPTED TEST RESULTS")
print("=" * 92)
display(
    PAD_DA_TEST_RESULTS_DF
)

print(
    "\nValidation-selected PAD adapted method:",
    PAD_DA_SELECTED_METHOD,
)

selected_test_row = (
    PAD_DA_TEST_RESULTS_DF.loc[
        PAD_DA_TEST_RESULTS_DF[
            "method"
        ] == PAD_DA_SELECTED_METHOD
    ].iloc[0]
)

print(
    "Selected method PAD test BACC:",
    f"{100.0 * selected_test_row['bacc']:.4f}%",
)
````
